Esta ferramenta simula a **dinâmica de uma partícula** de massa $m$ sob a ação de uma **força central** $\vec{F}=F_r(r)\hat{r}$ que aponta sempre ao longo da direção radial e cuja intensidade depende apenas da distância $r$ entre a partícula e a origem do sistema de coordenadas. A solução é obtida numericamente a partir do **problema de valor inicial (PVI)** associado à **2ª Lei de Newton**. O PVI é composto pelas equações diferenciais ordinárias (EDOs) e pelas condições iniciais do problema.


Toda força central é conservativa e a componente radial $F_r$ da força pode ser derivada a partir da energia potencial $U$, que só depende da coordenada radial. Você pode fornecer **qualquer uma das duas**:

- a **energia potencial** $U(r)$, e a ferramenta calcula analiticamente $F_r(r) = -\dfrac{dU}{dr}$; ou
- a **força radial** $F_r(r)$, e a ferramenta calcula analiticamente $U(r) = -\displaystyle\int F_r(r)\,dr$ (a constante aditiva, escolhida arbitrariamente, não afeta a dinâmica).

---
### Sistemas de coordenadas disponíveis

Como o momento angular $\vec{L}$ é sempre conservado em um problema de forças centrai, o movimento está **confinado a um plano** que passa pela origem e é perpendicular a $\vec{L}$. Por isso, esta ferramenta oferece apenas dois sistemas de coordenadas, que são os mais naturais para este tipo de problema:

- **Polares** $(\rho,\varphi)$: o movimento é forçado a ficar no plano $xy$;
- **Esféricas** $(r,\varphi,\theta)$: o movimento é calculado em 3D, mas a própria dinâmica confina a trajetória a um plano, que, dependendo das condições iniciais, pode estar inclinado em relação ao plano $xy$; se $\theta=\pi/2$ e $v_\theta=0$ fizerem parte das condições iniciais, este caso será equivalente ao anterior.

Nas coordenadas polares e esféricas, $\varphi$ é o ângulo azimutal (medido no plano $xy$); nas coordenadas esféricas, $\theta$ é o ângulo polar (medido a partir do semieixo $z$ positivo), sempre em **radianos**.


As coordenadas cartesianas são expressas, em termos das coordenadas polares e das coordenadas esféricas, por:

| Coordenadas polares | Coordenadas esféricas |
|:---|:---|
| $x=\rho\cos\varphi$   | $x=r\cos\varphi\sin\theta$   |
| $y=\rho\sin\varphi$ |  $y=r\sin\varphi\sin\theta$ |
| $z=0$ | $z=r\cos\theta$ |

A partir da energia potencial $U$, a ferramenta calcula **analiticamente** as componentes $F_i$ da força $\vec{F}$, usando o gradiente em coordenadas ortogonais curvilíneas:

$$F_i = -\frac{1}{h_i}\frac{\partial U}{\partial q_i}$$

com os fatores de escala $h_i$ adequados para cada sistema:

| Coordenadas | $F_1$ | $F_2$ | $F_3$ |
|:---|:---:|:---:|:---:|
| Cartesianas | $F_x=-\dfrac{\partial U}{\partial x}$ | $F_y=-\dfrac{\partial U}{\partial y}$ | $F_z=-\dfrac{\partial U}{\partial z}$ |
| Polares | $F_\rho=-\dfrac{\partial U}{\partial \rho}$ | $F_\varphi=-\dfrac{1}{\rho}\dfrac{\partial U}{\partial \varphi}$ | $F_z=0$ |
| Esféricas | $F_r=-\dfrac{\partial U}{\partial r}$ | $F_\varphi=-\dfrac{1}{r\,\mathrm{sen}\,\theta}\dfrac{\partial U}{\partial \varphi}$ | $F_\theta=-\dfrac{1}{r}\dfrac{\partial U}{\partial \theta}$ |

As expressões analíticas calculadas para $F_1,F_2,F_3$ são exibidas na interface antes da integração numérica. A energia potencial $U$ ou a componente radial $F_r$ da força podem depender apenas da massa e da coordenadas radial da partícula, não podendo depender explicitamente do tempo $t$, da velocidade $\vec{v}$ da partícula ou de coordenadas angulares.

As componentes das velocidade e da aceleração usadas em todos os campos e nos gráficos são as **componentes físicas** da velocidade (com dimensão $\mathrm{L}\mathrm{T}^{-1}$) ou da aceleração (com dimensão $\mathrm{L}\mathrm{T}^{-2}$) e não grandezas como as velocidades angulares $\dot{\varphi}$ e $\dot{\theta}$ ou as acelerações angulares $\ddot{\varphi}$ e $\ddot{\theta}$. ### Sintaxe das expressões de $F_r$ e $U$

As variáveis disponíveis são `m` (massa) e a coordenada radial do sistema escolhido (`rho` para Polares ou `r` para Esféricas). As expressões devem ser digitadas usando a notação do pacote SymPy (https://www.sympy.org/):

| Para obter &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp;| Digite &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; |
|:---|:---|
| $\tfrac12 k\rho^2$, com $k=20,\!0\,\mathrm{N/m}$ | `10.0*rho**2` |
| $-k\rho$, com $k=20,\!0\,\mathrm{N/m}$ | `-20.0*rho` |
| $-\dfrac{GM_\odot m}{r}$, com $GM_\odot=132,\!7\times 10^{18}\,\mathrm{m^3/s^2}$ | `-132.7e18*m/r` |
| $-\dfrac{GM_\odot m}{r^2}$, com $GM_\odot=132,\!7\times 10^{18}\,\mathrm{m^3/s^2}$ | `-132.7e18*m/r**2` |

1. **Constantes** (como $k$, $g$ ou $GM_\odot$) devem ser digitadas como **números**, não como símbolos soltos — a ferramenta não reconhece letras que não sejam `m` ou as coordenadas radias $\rho$ ou $r$ do sistema escolhido.
2. **Evite usar o símbolo `I`**, porque `I` é reservado pelo SymPy para representar a unidade imaginária $i=\sqrt{-1}$.
3. **Evite usar o símbolo `E`**, porque `E` é reservado pelo SymPy para representar a base do logaritmo natural $e=\exp(1)$.
4. Além disso, os símbolos `S`, `N`, `C`, `O`, `Q`, `zoo`, `nan`, `oo`, `true`, `false`, `True` e `False` não devem ser usados, porque também são símbolos reservados pelo SymPy.
5. Potências de `10` podem ser representadas pelo símbolo `e`: $3,\!14\times 10^{-6}$ pode ser escito como `3.14e-6`, por exemplo.

### O que a ferramenta calcula e mostra

Você informa a massa $m$, o sistema de coordenadas, a energia potencial $U$ ou a componente radial $F_r$ da força, o intervalo de simulação $[t_\mathrm{min},t_\mathrm{max}]$ e as condições iniciais (posição e velocidade em $t=t_\mathrm{min}$). A ferramenta calcula a grandeza que faltar ($F_r$ ou $U$) analiticamente, resolve numericamente as equações de movimento com `scipy.integrate.solve_ivp` (método e tolerâncias configuráveis nas opções avançadas) e mostra:

- a **energia potencial** $U(r)$, a **barreira centrífuga** $L^2/(2mr^2)$, o **potencial efetivo** $U_\mathrm{ef}(r)=U(r)+L^2/(2mr^2)$ e a **energia mecânica** $E$ (constante), todos em função da coordenada radial, e marca os pontos de retorno da simulação;
- as componentes $L_x,L_y,L_z$ e o módulo $|\vec L|$ do momento angular em função do tempo (em Coordenaddas Polares, mostra apenas $L_z$, pois $L_x=0=L_y$);
- a trajetória da partícula, colorida pela velocidade escalar $|\vec v(t)|$;
- as componentes de posição, velocidade e aceleração em função do tempo;
- a **órbita vista de cima**: um gráfico polar de $r$ em função do *ângulo no do plano do movimento*, também colorido pela velocidade.

O painel de informações indica a variação percentual da energia mecânica e do módulo do momento angular ao longo da simulação e, no caso 3D, o desvio máximo da trajetória em relação ao plano orbital. No caso ideal, as variações percentuais e os desvio máximo deveriam ser nulos.

---
**Atenção**: os campos de unidade de posição e de tempo servem apenas para rotular eixos e tabelas e **não realizam conversão entre unidades**. Cabe a você garantir que $m$, $U$ e $F_r$, as posições, velocidades e o tempo estejam todos em um sistema de unidades coerente, como o Sistema Internacional. Os ângulos $\varphi$ e $\theta$ são sempre tratados em radianos.

**Atenção**: a origem ($\rho=0$ ou $r=0$) é uma singularidade dos sistemas polares e esférico (o ângulo azimutal não é definido ali) — evite condições iniciais ou trajetórias que passem exatamente pela origem nesses sistemas. Da mesma forma, os polos ($\theta=0$ ou $\theta=\pi$) são singulares no sistema esférico.

```python
#@title Execute esta célula para abrir a ferramenta **NEWTON TOOLS &mdash; Forças Centrais**
# ── Célula 1: Forças Centrais (Problema de Valor Inicial) ───────────────────
# A ferramenta foi especificada pelo licenciante, incluindo modelos, métodos,
# pacotes a serem usados e linhas gerais da inteface gráfica, mas a maior parte
# do código, especialmente as funções para leitura de arquivos e para a
# interface gráfica, foi escrita usando o modelo Claude Sonnet 5.
# O código gerado foi inspecionado e sofreu alterações realizadas pelo licenciante.

import math, io, base64, warnings
import numpy as np
import sympy as sp
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Line3DCollection
from matplotlib.collections import LineCollection
from sympy.parsing.sympy_parser import (
    parse_expr, standard_transformations, implicit_multiplication_application
)
from scipy.integrate import solve_ivp
import ipywidgets as w
from IPython.display import display, HTML, Math, clear_output
import zipfile

warnings.filterwarnings("ignore")
matplotlib.rcParams.update({"figure.dpi": 110, "font.size": 10,
                             "axes.spines.top": False,
                             "axes.spines.right": False})

# Detecta se está rodando no Google Colab
try:
    from google.colab import files as colab_files
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# Transformações para aceitar entradas como "2*m" com multiplicação implícita
TRANSFORMATIONS = standard_transformations + (implicit_multiplication_application,)

# ══════════════════════════════════════════════════════════════════
#  ESTILOS CSS
#  Convenção compartilhada por toda a suíte LAB/NEWTON TOOLS:
#    .sec   → rótulo de seção (cor = acento do app)
#    .rcard → card de resultado/destaque (cor = acento do app)
#    .gcard → card de "sucesso" (verde fixo)
#    .bcard → card informativo secundário (azul fixo)
#    .wcard → card de aviso (laranja fixo)
#    .ecard → card de erro (vermelho fixo)
#    .icard → card informativo auxiliar (violeta fixo)
#    .res-t → tabela de dados (cabeçalho azul-marinho fixo)
#  Acento desta ferramenta: ciano #1098ad
# ══════════════════════════════════════════════════════════════════
display(HTML("""
<style>
.sec{font-weight:700;font-size:.75rem;letter-spacing:.08em;color:#1098ad;
  text-transform:uppercase;margin:13px 0 3px;}
.rcard{background:#eafcff;border-left:4px solid #1098ad;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-family:monospace;font-size:.91rem;}
.gcard{background:#f0fff4;border-left:4px solid #2f9e44;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.91rem;}
.bcard{background:#f0f4ff;border-left:4px solid #3b5bdb;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-family:monospace;font-size:.91rem;}
.wcard{background:#fff9db;border-left:4px solid #e67700;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.91rem;}
.ecard{background:#fff0f0;border-left:4px solid #c92a2a;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.91rem;}
.icard{background:#f3f0ff;border-left:4px solid #7048e8;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.88rem;}
.res-t{border-collapse:collapse;font-size:.83rem;width:60%;margin:6px 0;}
.res-t th{background:#1a1a2e;color:#fff;padding:5px 10px;text-align:left;}
.res-t td{padding:4px 10px;text-align:left;border-bottom:1px solid #eee;
  font-family:monospace;}
.res-t tr:nth-child(odd) td{text-align:left}
.res-t tr:nth-child(even) td{background:#eafcff;text-align:left}
</style>
"""))

# ══════════════════════════════════════════════════════════════════
#  HTML HELPERS
# ══════════════════════════════════════════════════════════════════
def h(tag, content, **attrs):
    a = " ".join(f'{k.rstrip("_")}="{v}"' for k, v in attrs.items())
    return f"<{tag} {a}>{content}</{tag}>"

def card(s, c="rcard"): return f'<div class="{c}">{s}</div>'

def sec_html(texto, estilo=""):
    """Constrói o HTML de um rótulo de seção (classe .sec)."""
    extra = f' style="{estilo}"' if estilo else ""
    return f'<div class="sec"{extra}>{texto}</div>'

def titulo_secao(texto, estilo=""):
    """Widget HTML com um rótulo de seção — forma padrão de criar
    títulos de seção em toda a suíte LAB/NEWTON TOOLS."""
    return w.HTML(sec_html(texto, estilo))

def tabela_html(df):
    """Renderiza um DataFrame com o mesmo estilo .res-t usado nas
    demais ferramentas da suíte."""
    return df.to_html(classes="res-t", index=False, border=0,
                       float_format=lambda v: f"{v:.6g}")

def fig_para_html(fig, max_width=900):
    """Converte uma figura matplotlib em <img> embutido (PNG base64),
    no mesmo padrão visual usado nas demais ferramentas da suíte.
    Não fecha a figura — ela é reutilizada na exportação em PNG."""
    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=130, bbox_inches="tight")
    buf.seek(0)
    b64 = base64.b64encode(buf.read()).decode()
    return (f'<img src="data:image/png;base64,{b64}" '
            f'style="width:100%;max-width:{max_width}px;border-radius:6px;'
            f'box-shadow:0 2px 8px rgba(0,0,0,.13)">')

# ══════════════════════════════════════════════════════════════════
#  MOTOR FÍSICO — força central: F = F_r(r)·ê_r (só depende da
#  distância à origem; nunca tem componente tangencial). O estudante
#  fornece U(r) ou F_r(r); a ferramenta calcula analiticamente a outra.
#
#  Dois sistemas de coordenadas, com dimensões diferentes:
#    Polares  (ρ, φ)        — movimento sempre no plano xy (2 graus de liberdade)
#    Esféricas (r, φ, θ)    — movimento 3D, mas confinado a um plano fixo
#                              que contém a origem (teorema da força central)
#
#  Convenção de velocidades: sempre as componentes FÍSICAS (ortonormais),
#  com dimensão de [posição]/[tempo] — v_φ = ρ·dφ/dt (polares) ou
#  v_φ = r·sen(θ)·dφ/dt e v_θ = r·dθ/dt (esféricas). Isso garante que
#  K = (1/2) m (v1²+v2²+ ... ) valha sempre.
# ══════════════════════════════════════════════════════════════════

m_sym = sp.symbols('m', positive=True)

COORD_DEFS = {
    'Polares': dict(
        key='polar',
        dim=2,
        titulo='Polares (ρ, φ)',
        q_syms=['rho', 'phi'], v_syms=['vrho', 'vphi'],
        q_disp=['ρ', 'φ'], v_disp=['v_ρ', 'v_φ'],
        a_disp=['a_ρ', 'a_φ'],
        q_kind=['pos', 'ang'],
        q_default=[1.5, 0.0], v_default=[0.0, 1.3],
        U_default='-4*m/rho', Fr_default='-4*m/rho**2',
        obs=('Movimento sempre confinado ao plano xy. ρ é a distância à '
             'origem e φ é o ângulo azimutal (radianos). '
             'v_φ = ρ·dφ/dt é a componente FÍSICA (tangencial) da '
             'velocidade — não confundir com dφ/dt.'),
    ),
    'Esféricas': dict(
        key='sph',
        dim=3,
        titulo='Esféricas (r, φ, θ)',
        q_syms=['r', 'phi', 'theta'], v_syms=['vr', 'vphi', 'vtheta'],
        q_disp=['r', 'φ', 'θ'], v_disp=['v_r', 'v_φ', 'v_θ'],
        a_disp=['a_r', 'a_φ', 'a_θ'],
        q_kind=['pos', 'ang', 'ang'],
        q_default=[1.5, 0.2, 1.1], v_default=[0.0, 1.1, 0.3],
        U_default='-4*m/r', Fr_default='-4*m/r**2',
        obs=('Embora o movimento seja calculado em 3D, uma força central '
             'sempre confina a trajetória a um plano fixo que passa pela '
             'origem (perpendicular ao momento angular). r é a distância à '
             'origem, φ é o ângulo azimutal e θ é o ângulo polar (a partir '
             'de +z), ambos em radianos. v_φ = r·sen(θ)·dφ/dt e '
             'v_θ = r·dθ/dt são as componentes FÍSICAS da velocidade.'),
    ),
}

RESERVED = {'I', 'E', 'S', 'N', 'C', 'O', 'Q', 'zoo', 'nan', 'oo',
            'true', 'false', 'True', 'False'}

EPS = 1e-9


def local_dict_radial(coord_key_name):
    """Símbolos SymPy (m, q1) — a coordenada radial do sistema escolhido
    (ρ para Polares, r para Esféricas). U e F_r só podem depender dela."""
    cfg = COORD_DEFS[coord_key_name]
    q1 = sp.symbols(cfg['q_syms'][0], positive=True)
    ld = {'m': m_sym, cfg['q_syms'][0]: q1}
    return ld, q1


def parse_radial(texto, local_dict, q1_sym, rotulo):
    """Interpreta a expressão digitada para U(r) ou F_r(r): só pode
    depender de m e da coordenada radial — nunca de t, ângulos ou
    velocidades (é isso que caracteriza uma força CENTRAL)."""
    expr = parse_expr(texto, local_dict=local_dict, transformations=TRANSFORMATIONS)
    usados = {str(s) for s in expr.free_symbols}
    bad_reserved = usados & RESERVED
    if bad_reserved:
        raise ValueError(f"símbolo(s) reservado(s) do SymPy usados: {', '.join(sorted(bad_reserved))}")
    allowed = {'m', str(q1_sym)}
    bad_unknown = usados - allowed
    if bad_unknown:
        raise ValueError(
            f"símbolo(s) não reconhecido(s): {', '.join(sorted(bad_unknown))}. "
            f"Uma força CENTRAL só pode depender de m e de {q1_sym} — "
            f"{rotulo} nunca depende do tempo, de ângulos ou de velocidades; "
            f"constantes devem ser digitadas como números.")
    return expr


def make_num_funcs(expr, arg_syms):
    """Retorna (f_scalar, f_vec) para uma expressão simbólica, avaliando-a
    em argumentos escalares (integrador) ou vetoriais/numpy (gráficos)."""
    if expr.free_symbols == set():
        val = float(expr)
        def f_scalar(*a):
            return val
        def f_vec(*a):
            ref = np.asarray(a[0], dtype=float)
            return np.full_like(ref, val)
        return f_scalar, f_vec

    f_np = sp.lambdify(arg_syms, expr, 'numpy')

    def f_scalar(*a):
        return float(f_np(*a))

    def f_vec(*a):
        y = np.asarray(f_np(*a), dtype=float)
        ref = np.asarray(a[0], dtype=float)
        if y.shape != ref.shape:
            y = np.full_like(ref, float(y))
        return y

    return f_scalar, f_vec


# ---------------------------------------------------------------------
# Equações de movimento (2ª Lei de Newton) para uma força puramente
# central (F_φ = F_θ = 0 sempre). Reduções das equações gerais em
# coordenadas curvilíneas, validadas numericamente contra integração
# cartesiana direta.
# ---------------------------------------------------------------------

def make_rhs_polar(massa, Fr_scalar):
    """Estado y = [rho, phi, vrho, vphi] (2 graus de liberdade)."""
    def rhs(tt, y):
        rho, phi, vrho, vphi = y
        rho_s = rho if abs(rho) > EPS else (EPS if rho >= 0 else -EPS)
        Fr = Fr_scalar(massa, rho)
        drho = vrho
        dphi = vphi / rho_s
        dvrho = Fr / massa + vphi**2 / rho_s
        dvphi = -vrho * vphi / rho_s
        return [drho, dphi, dvrho, dvphi]
    return rhs


def make_rhs_sph(massa, Fr_scalar):
    """Estado y = [r, phi, theta, vr, vphi, vtheta] (3D, mas planar)."""
    def rhs(tt, y):
        r, phi, theta, vr, vphi, vtheta = y
        r_s = r if abs(r) > EPS else (EPS if r >= 0 else -EPS)
        sin_t = math.sin(theta)
        sin_s = sin_t if abs(sin_t) > EPS else (EPS if sin_t >= 0 else -EPS)
        cot_t = math.cos(theta) / sin_s
        Fr = Fr_scalar(massa, r)
        dr = vr
        dphi = vphi / (r_s * sin_s)
        dtheta = vtheta / r_s
        dvr = Fr / massa + (vtheta**2 + vphi**2) / r_s
        dvphi = -vr * vphi / r_s - vtheta * vphi * cot_t / r_s
        dvtheta = -vr * vtheta / r_s + vphi**2 * cot_t / r_s
        return [dr, dphi, dtheta, dvr, dvphi, dvtheta]
    return rhs


def calc_aceleracoes_polar(massa, Fr_vec, ts, rho, vrho, vphi):
    Frv = Fr_vec(np.full_like(ts, massa), rho)
    rho_s = np.where(np.abs(rho) > EPS, rho, EPS)
    a1 = Frv / massa + vphi**2 / rho_s
    a2 = -vrho * vphi / rho_s
    return a1, a2


def calc_aceleracoes_sph(massa, Fr_vec, ts, r, theta, vr, vphi, vtheta):
    Frv = Fr_vec(np.full_like(ts, massa), r)
    r_s = np.where(np.abs(r) > EPS, r, EPS)
    sin_t = np.sin(theta)
    sin_s = np.where(np.abs(sin_t) > EPS, sin_t, EPS)
    cot_t = np.cos(theta) / sin_s
    a1 = Frv / massa + (vtheta**2 + vphi**2) / r_s
    a2 = -vr * vphi / r_s - vtheta * vphi * cot_t / r_s
    a3 = -vr * vtheta / r_s + vphi**2 * cot_t / r_s
    return a1, a2, a3


def para_cartesianas_polar(rho, phi, vrho, vphi):
    x = rho * np.cos(phi)
    y = rho * np.sin(phi)
    vx = vrho * np.cos(phi) - vphi * np.sin(phi)
    vy = vrho * np.sin(phi) + vphi * np.cos(phi)
    return x, y, vx, vy


def para_cartesianas_sph(r, phi, theta, vr, vphi, vtheta):
    sin_t, cos_t = np.sin(theta), np.cos(theta)
    sin_p, cos_p = np.sin(phi), np.cos(phi)
    x = r * sin_t * cos_p
    y = r * sin_t * sin_p
    z = r * cos_t
    vx = vr * sin_t * cos_p + vtheta * cos_t * cos_p - vphi * sin_p
    vy = vr * sin_t * sin_p + vtheta * cos_t * sin_p + vphi * cos_p
    vz = vr * cos_t - vtheta * sin_t
    return x, y, z, vx, vy, vz

# ----------------------- Cabeçalho -----------------------

cabecalho = w.HTML(r'''
<div style="background: linear-gradient(135deg, #004587 0%, #00AB45 80%, #006EAB 100%);
            padding: 14px 20px; border-radius: 10px; color: white;
            font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 12px;
            box-shadow: 0 2px 6px rgba(0,0,0,0.15);">
  <div style="font-size: 13px; letter-spacing: 3px; opacity: 0.85; font-weight: 600;">QapybaLAB &mdash; NEWTON TOOLS</div>
  <div style="font-size: 26px; font-weight: 700; margin-top: 2px;">🎯 Dinâmica da Partícula - Forças Centrais</div>
</div>
''')

# ----------------------- Widgets de entrada -----------------------

massa_text = w.FloatText(
    value=1.0, description='Massa m (kg):',
    style={'description_width': 'initial'}, layout=w.Layout(width='220px')
)

coord_dropdown = w.Dropdown(
    options=list(COORD_DEFS.keys()), value='Polares',
    description='Sistema de coordenadas:',
    style={'description_width': 'initial'}, layout=w.Layout(width='320px')
)

obs_coord_html = w.HTML()

entrada_dropdown = w.Dropdown(
    options=['Energia potencial U(r)', 'Força radial F_r(r)'],
    value='Energia potencial U(r)',
    description='Fornecer:',
    style={'description_width': 'initial'}, layout=w.Layout(width='320px')
)

entrada_text = w.Text(layout=w.Layout(width='420px'), style={'description_width': '90px'})

tmin_text = w.FloatText(value=0.0, description='t mínimo:',
                         style={'description_width': 'initial'}, layout=w.Layout(width='170px'))
tmax_text = w.FloatText(value=8.0, description='t máximo:',
                         style={'description_width': 'initial'}, layout=w.Layout(width='170px'))

q1_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
q2_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
q3_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
v1_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
v2_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
v3_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})

linha_q3 = w.HBox([q3_0])
linha_v3 = w.HBox([v3_0])

unidade_pos = w.Text(value='m', description='Unidade de posição:',
                      style={'description_width': 'initial'}, layout=w.Layout(width='200px'))
unidade_tempo = w.Text(value='s', description='Unidade de tempo:',
                        style={'description_width': 'initial'}, layout=w.Layout(width='200px'))

unidades_obs = w.HTML(card(
    '⚠ Os campos de unidade servem apenas para rotular eixos e tabelas — '
    'não convertem valores. Garanta que m, U ou F_r, posições e tempo '
    'estejam todos em um sistema de unidades coerente (recomendado: '
    'SI &mdash; kg, m, s, J, N).',
    'wcard'))

# --- opções numéricas avançadas ---
metodo_dropdown = w.Dropdown(
    options=['RK45', 'RK23', 'DOP853', 'Radau', 'BDF', 'LSODA'],
    value='RK45', description='Método do integrador:',
    style={'description_width': 'initial'}, layout=w.Layout(width='260px')
)
npts_text = w.IntText(value=600, description='Nº de pontos para os gráficos:',
                       style={'description_width': 'initial'}, layout=w.Layout(width='260px'))
rtol_text = w.FloatText(value=1e-8, description='Tolerância relativa (rtol):',
                         style={'description_width': 'initial'}, layout=w.Layout(width='260px'))
atol_text = w.FloatText(value=1e-10, description='Tolerância absoluta (atol):',
                         style={'description_width': 'initial'}, layout=w.Layout(width='260px'))

opcoes_avancadas = w.Accordion(children=[w.VBox([
    w.HBox([metodo_dropdown, npts_text]),
    w.HBox([rtol_text, atol_text]),
])])
opcoes_avancadas.set_title(0, '⚙️ Opções numéricas (avançado)')
opcoes_avancadas.selected_index = None  # inicia recolhido

simular_btn = w.Button(description='▶  Simular', button_style='primary', icon='rocket',
                        layout=w.Layout(width='160px', height='36px'))

export_png_btn = w.Button(description='Gráficos (PNG)', icon='image')
export_xlsx_btn = w.Button(description='Exportar XLSX', icon='file-excel')
export_csv_btn = w.Button(description='Exportar CSV', icon='file-text')

out_info = w.Output()
out_energia = w.Output()
out_traj = w.Output()
out_grid = w.Output()
out_orbita = w.Output()
out_export = w.Output()


def atualizar_coordenadas(change=None):
    cfg = COORD_DEFS[coord_dropdown.value]

    if entrada_dropdown.value.startswith('Energia'):
        entrada_text.description = f"U({cfg['q_syms'][0]}) ="
        entrada_text.value = cfg['U_default']
    else:
        entrada_text.description = f"F_r({cfg['q_syms'][0]}) ="
        entrada_text.value = cfg['Fr_default']

    q1_0.description = f"{cfg['q_disp'][0]}₀ ="
    q2_0.description = f"{cfg['q_disp'][1]}₀ ="
    v1_0.description = f"{cfg['v_disp'][0]}₀ ="
    v2_0.description = f"{cfg['v_disp'][1]}₀ ="
    q1_0.value, q2_0.value = cfg['q_default'][0], cfg['q_default'][1]
    v1_0.value, v2_0.value = cfg['v_default'][0], cfg['v_default'][1]

    if cfg['dim'] == 3:
        q3_0.description = f"{cfg['q_disp'][2]}₀ ="
        v3_0.description = f"{cfg['v_disp'][2]}₀ ="
        q3_0.value, v3_0.value = cfg['q_default'][2], cfg['v_default'][2]
        linha_q3.layout.display = ''
        linha_v3.layout.display = ''
    else:
        linha_q3.layout.display = 'none'
        linha_v3.layout.display = 'none'

    unid_ang_txt = ' (φ, θ sempre em radianos)' if 'ang' in cfg['q_kind'] else ''
    obs_coord_html.value = card(cfg['obs'] + unid_ang_txt, 'icard')


coord_dropdown.observe(atualizar_coordenadas, names='value')
entrada_dropdown.observe(atualizar_coordenadas, names='value')
atualizar_coordenadas()

# ----------------------- Lógica principal -----------------------

state = {}

COR_POS = '#1565c0'
COR_VEL = '#ef6c00'
COR_ACE = '#2e7d32'
COR_U = '#e8590c'
COR_BARR = '#9c36b5'
COR_UEFF = '#0d47a1'
COR_E = '#1a1a2e'
COR_LX = '#c92a2a'
COR_LY = '#2f9e44'
COR_LZ = '#1565c0'
COR_LMAG = '#1a1a2e'


def on_simular_clicked(btn=None):
    for o in (out_info, out_energia, out_traj, out_grid, out_orbita, out_export):
        o.clear_output()

    massa = massa_text.value
    if not np.isfinite(massa) or massa <= 0:
        with out_info:
            display(HTML(card('✘ A massa m deve ser um número positivo.', 'ecard')))
        return

    tmin, tmax = tmin_text.value, tmax_text.value
    if tmin >= tmax:
        with out_info:
            display(HTML(card('✘ O tempo mínimo deve ser menor que o tempo máximo.', 'ecard')))
        return

    npts = max(int(npts_text.value), 20)
    rtol, atol = rtol_text.value, atol_text.value
    metodo = metodo_dropdown.value

    coord_name = coord_dropdown.value
    cfg = COORD_DEFS[coord_name]
    coord_key = cfg['key']
    dim = cfg['dim']
    up = unidade_pos.value.strip() or 'm'
    ut = unidade_tempo.value.strip() or 's'

    ld, q1_sym = local_dict_radial(coord_name)

    # --- Interpreta U(r) ou F_r(r), calculando analiticamente a outra ---
    fornecendo_U = entrada_dropdown.value.startswith('Energia')
    try:
        if fornecendo_U:
            U_expr = parse_radial(entrada_text.value, ld, q1_sym, 'a energia potencial U(r)')
            Fr_expr = sp.simplify(-sp.diff(U_expr, q1_sym))
        else:
            Fr_expr = parse_radial(entrada_text.value, ld, q1_sym, 'a força radial F_r(r)')
            integral = sp.integrate(Fr_expr, q1_sym)
            if integral.has(sp.Integral):
                raise ValueError('não foi possível calcular ∫F_r(r)dr analiticamente; '
                                  'tente uma expressão mais simples ou forneça U(r) diretamente.')
            U_expr = sp.simplify(-integral)
    except Exception as e:
        rotulo = 'U(r)' if fornecendo_U else 'F_r(r)'
        with out_info:
            display(HTML(card(f'✘ Erro ao interpretar {rotulo}: {e}', 'ecard')))
        return

    Fr_scalar, Fr_vec = make_num_funcs(Fr_expr, (m_sym, q1_sym))
    U_scalar, U_vec = make_num_funcs(U_expr, (m_sym, q1_sym))

    # --- Condições iniciais ---
    q0 = [q1_0.value, q2_0.value] + ([q3_0.value] if dim == 3 else [])
    v0 = [v1_0.value, v2_0.value] + ([v3_0.value] if dim == 3 else [])

    avisos = []

    nome0 = cfg['q_disp'][0]
    if abs(q0[0]) < 1e-6:
        with out_info:
            display(HTML(card(
                f'✘ A origem ({nome0} = 0) é uma singularidade do sistema de '
                f'coordenadas {cfg["titulo"]} (o ângulo azimutal não é definido '
                f'ali). Escolha {nome0}₀ > 0.',
                'ecard')))
        return

    if coord_key == 'sph' and abs(math.sin(q0[2])) < 1e-6:
        avisos.append(
            'A condição inicial está muito próxima de um polo (θ₀ ≈ 0 ou π), '
            'onde o ângulo azimutal φ é indefinido. Os resultados podem ficar '
            'numericamente instáveis nessa região.')

    # momento angular inicial (magnitude) — usado na barreira centrífuga
    if coord_key == 'polar':
        L0 = massa * q0[0] * v0[1]
    else:
        x0c, y0c, z0c, vx0c, vy0c, vz0c = para_cartesianas_sph(
            q0[0], q0[1], q0[2], v0[0], v0[1], v0[2])
        L0 = massa * math.sqrt((y0c*vz0c - z0c*vy0c)**2 +
                                (z0c*vx0c - x0c*vz0c)**2 +
                                (x0c*vy0c - y0c*vx0c)**2)

    if abs(L0) < 1e-6 * max(massa * q0[0] * max(abs(v0[0]), abs(v0[1]), 1e-9), 1e-9):
        avisos.append(
            'O momento angular inicial é (quase) nulo — a velocidade inicial é '
            'quase puramente radial. A partícula pode cair na origem (colisão '
            'radial), onde o sistema de coordenadas é singular.')

    # --- Integração numérica ---
    if coord_key == 'polar':
        rhs = make_rhs_polar(massa, Fr_scalar)
    else:
        rhs = make_rhs_sph(massa, Fr_scalar)

    t_eval = np.linspace(tmin, tmax, npts)
    try:
        sol = solve_ivp(rhs, (tmin, tmax), q0 + v0, t_eval=t_eval,
                         method=metodo, rtol=rtol, atol=atol)
    except Exception as e:
        with out_info:
            display(HTML(card(f'✘ Erro durante a integração numérica: {e}', 'ecard')))
        return

    if not sol.success:
        with out_info:
            display(HTML(card(f'✘ O integrador não convergiu: {sol.message}', 'ecard')))
        return

    ts = sol.t

    if coord_key == 'polar':
        q1, q2, v1, v2 = sol.y
        a1, a2 = calc_aceleracoes_polar(massa, Fr_vec, ts, q1, v1, v2)
        x, y, vx, vy = para_cartesianas_polar(q1, q2, v1, v2)
        z = np.zeros_like(x)
        vz = np.zeros_like(vx)
    else:
        q1, q2, q3, v1, v2, v3 = sol.y
        a1, a2, a3 = calc_aceleracoes_sph(massa, Fr_vec, ts, q1, q3, v1, v2, v3)
        x, y, z, vx, vy, vz = para_cartesianas_sph(q1, q2, q3, v1, v2, v3)

    if np.min(np.abs(q1)) < 1e-4 * max(abs(q0[0]), 1.0):
        avisos.append(
            f'A trajetória passou muito perto da origem ({nome0} ≈ 0) durante '
            'a simulação — região singular do sistema de coordenadas. '
            'Verifique os resultados com cautela.')

    K = 0.5 * massa * (v1**2 + v2**2 + (v3**2 if coord_key == 'sph' else 0.0))
    U_t = U_vec(np.full_like(ts, massa), q1)
    Emec = K + U_t

    Lx = massa * (y * vz - z * vy)
    Ly = massa * (z * vx - x * vz)
    Lz = massa * (x * vy - y * vx)
    Lmag = np.sqrt(Lx**2 + Ly**2 + Lz**2)

    E0 = Emec[0]
    drift_E = (np.max(Emec) - np.min(Emec)) / max(abs(E0), 1e-12)
    drift_L = (np.max(Lmag) - np.min(Lmag)) / max(abs(Lmag[0]), 1e-12)
    if drift_E > 1e-2 or drift_L > 1e-2:
        avisos.append(
            f'A energia mecânica variou cerca de {100*drift_E:.2g}% e o módulo '
            f'do momento angular cerca de {100*drift_L:.2g}% ao longo da '
            'simulação. Para uma força central, ambos deveriam permanecer '
            'praticamente constantes — considere reduzir rtol/atol ou trocar '
            'o método do integrador nas opções avançadas.')

    # --- Verificação de planaridade (só é uma checagem não trivial em 3D) ---
    desvio_plano_pct = None
    if coord_key == 'sph' and abs(Lmag[0]) > 1e-9:
        Lhat = np.array([Lx[0], Ly[0], Lz[0]]) / Lmag[0]
        r_vec = np.stack([x, y, z], axis=1)
        fora_do_plano = r_vec @ Lhat
        desvio_plano_pct = 100 * np.max(np.abs(fora_do_plano)) / max(np.mean(q1), 1e-9)

    state.clear()
    state.update(
        massa=massa, coord_name=coord_name, coord_key=coord_key, dim=dim,
        fornecendo_U=fornecendo_U, U_expr=str(U_expr), Fr_expr=str(Fr_expr),
        q0=q0, v0=v0, tmin=tmin, tmax=tmax, metodo=metodo, rtol=rtol, atol=atol,
        npts=npts, up=up, ut=ut, ts=ts, q1=q1, q2=q2, v1=v1, v2=v2,
        q3=(q3 if coord_key == 'sph' else None), v3=(v3 if coord_key == 'sph' else None),
        a1=a1, a2=a2, a3=(a3 if coord_key == 'sph' else None),
        x=x, y=y, z=z, vx=vx, vy=vy, vz=vz,
        K=K, U=U_t, Emec=Emec, Lx=Lx, Ly=Ly, Lz=Lz, Lmag=Lmag, L0=L0,
    )

    # --- Painel de informações ---
    with out_info:
        partes = []
        partes.append(sec_html('Energia potencial e força central', 'margin-top:0'))
        partes.append(card(
            f'U({q1_sym}) = {U_expr} &nbsp;&nbsp;|&nbsp;&nbsp; F_r({q1_sym}) = {Fr_expr}'
            + (' &nbsp; <i>(calculada a partir de U)</i>' if fornecendo_U
               else ' &nbsp; <i>(U calculada a partir de F_r; definida a menos de uma constante aditiva)</i>'),
            'rcard'))
        partes.append(sec_html('Condições iniciais e intervalo'))
        nomes_q = cfg['q_disp']
        nomes_v = cfg['v_disp']
        ic_rows = ''.join(
            f'<tr><td>{nomes_q[i]}₀</td><td>{q0[i]:.6g} {up}'
            f'{" (rad)" if cfg["q_kind"][i]=="ang" else ""}</td>'
            f'<td>{nomes_v[i]}₀</td><td>{v0[i]:.6g} {up}/{ut}</td></tr>'
            for i in range(dim)
        )
        partes.append(
            f'<table class="res-t"><tr><th>Coordenada</th><th>Valor</th>'
            f'<th>Velocidade</th><th>Valor</th></tr>{ic_rows}</table>')
        partes.append(card(
            f'Massa m = {massa:.6g} kg &nbsp;|&nbsp; '
            f'Intervalo: t ∈ [{tmin:.6g}, {tmax:.6g}] {ut} &nbsp;|&nbsp; '
            f'método: {metodo} &nbsp;|&nbsp; pontos: {npts} &nbsp;|&nbsp; '
            f'rtol={rtol:g}, atol={atol:g} &nbsp;|&nbsp; '
            f'passos internos: {len(sol.t)} (nfev={sol.nfev})',
            'bcard'))
        info_extra = (f'Variação da energia mecânica: {100*drift_E:.3g}% &nbsp;|&nbsp; '
                      f'Variação do módulo do momento angular: {100*drift_L:.3g}%')
        if desvio_plano_pct is not None:
            info_extra += (f' &nbsp;|&nbsp; Desvio máximo do plano orbital: '
                            f'{desvio_plano_pct:.2g}% do raio típico '
                            f'(deveria ser ≈0 — confirma que o movimento é plano)')
        partes.append(card(info_extra, 'bcard'))
        for a in avisos:
            partes.append(card('⚠ ' + a, 'wcard'))
        display(HTML(''.join(partes)))

    # --- Figura A: energia (U, barreira, U_eff, E) vs r  +  momento angular vs t ---
    r_sim_min, r_sim_max = q1.min(), q1.max()
    span = max(r_sim_max - r_sim_min, 0.3 * q1.mean(), 1e-3)
    r_plot_min = max(1e-3, 0.75 * r_sim_min)
    r_plot_max = r_sim_max + 0.4 * span
    r_grid = np.linspace(r_plot_min, r_plot_max, 400)
    U_curve = U_vec(np.full_like(r_grid, massa), r_grid)
    barreira = L0**2 / (2 * massa * r_grid**2)
    U_eff = U_curve + barreira
    E_const = np.mean(Emec)

    fig_en, axe = plt.subplots(1, 2, figsize=(13, 4.6))

    axe[0].plot(r_grid, U_curve, color=COR_U, lw=1.8, label='U(r)')
    axe[0].plot(r_grid, barreira, color=COR_BARR, lw=1.8, ls=':', label='L²/(2mr²)  (barreira centrífuga)')
    axe[0].plot(r_grid, U_eff, color=COR_UEFF, lw=2.2, label='U_ef(r) = U(r) + L²/(2mr²)')
    axe[0].axhline(E_const, color=COR_E, lw=1.8, ls='--', label='E (mecânica)')
    axe[0].axvline(r_sim_min, color='0.5', lw=1.0, ls=':')
    axe[0].axvline(r_sim_max, color='0.5', lw=1.0, ls=':')
    U_eff_finite = U_eff[np.isfinite(U_eff)]
    U_eff_min = U_eff_finite.min() if U_eff_finite.size else E_const
    y_lo = min(U_curve.min(), U_eff_min) - 0.1 * max(E_const - U_eff_min, 0.1 * abs(E_const) + 0.1)
    y_hi = 1.1 * max(barreira.max(), U_eff_finite.max() if U_eff_finite.size else E_const, E_const)
    axe[0].set_ylim(y_lo, y_hi)
    axe[0].set_xlabel(f'{cfg["q_disp"][0]}  [{up}]')
    axe[0].set_ylabel('Energia  [J, assumindo SI]')
    axe[0].set_title('A) Energia potencial, barreira centrífuga,\npotencial efetivo e energia mecânica',
                      fontsize=10, loc='left', color=COR_UEFF)
    axe[0].legend(fontsize=7.5, loc='best')
    axe[0].grid(True, alpha=0.3)

    def _ylim_estavel(ax, *dados, frac=0.06):
        """Evita que ruído numérico de uma grandeza essencialmente constante
        (variação relativa muito pequena) seja ampliado pela escala
        automática do matplotlib a ponto de parecer uma oscilação real."""
        todos = np.concatenate(dados)
        dmed = np.mean(todos)
        escala = max(abs(dmed), np.max(np.abs(todos)), 1e-12)
        if (todos.max() - todos.min()) < 1e-4 * escala:
            margem = max(frac * escala, 1e-12)
            ax.set_ylim(dmed - margem, dmed + margem)

    if coord_key == 'sph':
        axe[1].plot(ts, Lx, color=COR_LX, lw=1.4, label='L_x(t)')
        axe[1].plot(ts, Ly, color=COR_LY, lw=1.4, label='L_y(t)')
        axe[1].plot(ts, Lz, color=COR_LZ, lw=1.4, label='L_z(t)')
        axe[1].plot(ts, Lmag, color=COR_LMAG, lw=2.2, ls='--', label='|L(t)|')
        _ylim_estavel(axe[1], Lx, Ly, Lz, Lmag)
    else:
        axe[1].plot(ts, Lz, color=COR_LMAG, lw=2.0, label='L(t)  (= L_z, movimento no plano xy)')
        _ylim_estavel(axe[1], Lz)
    axe[1].set_xlabel(f't  [{ut}]')
    axe[1].set_ylabel('Momento angular  [kg·{}²/{}]'.format(up, ut))
    axe[1].set_title('A) Componentes e módulo do momento angular',
                      fontsize=10, loc='left', color=COR_LMAG)
    axe[1].legend(fontsize=8)
    axe[1].grid(True, alpha=0.3)

    fig_en.suptitle('Diagrama de energia efetiva e conservação do momento angular',
                     fontsize=12, fontweight='bold', color='#0d47a1')
    fig_en.tight_layout(rect=[0, 0, 1, 0.92])
    state['fig_en'] = fig_en
    with out_energia:
        display(HTML(fig_para_html(fig_en, max_width=1150)))
    plt.close(fig_en)

    # --- Figura B: trajetória colorida pela velocidade (xy se Polares, 3D se Esféricas) ---
    speed = np.sqrt(v1**2 + v2**2 + (v3**2 if coord_key == 'sph' else 0.0))

    if coord_key == 'polar':
        fig_traj = plt.figure(figsize=(7.2, 6.6))
        axt = fig_traj.add_subplot(111)
        pontos = np.array([x, y]).T.reshape(-1, 1, 2)
        segmentos = np.concatenate([pontos[:-1], pontos[1:]], axis=1)
        lc = LineCollection(segmentos, cmap='viridis',
                             norm=plt.Normalize(speed.min(), speed.max()))
        lc.set_array(speed[:-1])
        lc.set_linewidth(3.0)
        axt.add_collection(lc)
        axt.scatter([0], [0], color='black', marker='+', s=90, zorder=6, label='origem')
        axt.scatter([x[0]], [y[0]], color='#2f9e44', s=60, zorder=5, label='Início')
        axt.scatter([x[-1]], [y[-1]], color='#c92a2a', s=60, zorder=5, label='Fim')
        pad = 0.08 * max(np.ptp(x), np.ptp(y), 1e-6)
        axt.set_xlim(x.min() - pad, x.max() + pad)
        axt.set_ylim(y.min() - pad, y.max() + pad)
        axt.set_xlabel(f'x [{up}]')
        axt.set_ylabel(f'y [{up}]')
        axt.set_aspect('equal', adjustable='box')
        axt.legend(loc='upper left', fontsize=8)
        cbar = fig_traj.colorbar(lc, ax=axt, shrink=0.75, pad=0.04)
        cbar.set_label(f'|v|  [{up}/{ut}]')
    else:
        fig_traj = plt.figure(figsize=(7.8, 6.6))
        axt = fig_traj.add_subplot(111, projection='3d')
        pontos = np.array([x, y, z]).T.reshape(-1, 1, 3)
        segmentos = np.concatenate([pontos[:-1], pontos[1:]], axis=1)
        lc = Line3DCollection(segmentos, cmap='viridis',
                               norm=plt.Normalize(speed.min(), speed.max()))
        lc.set_array(speed[:-1])
        lc.set_linewidth(3.0)
        axt.add_collection3d(lc)
        axt.scatter([x[0]], [y[0]], [z[0]], color='#2f9e44', s=55, label='Início', depthshade=False, zorder=5)
        axt.scatter([x[-1]], [y[-1]], [z[-1]], color='#c92a2a', s=55, label='Fim', depthshade=False, zorder=5)
        pad = 0.08 * max(np.ptp(x), np.ptp(y), np.ptp(z), 1e-6)
        axt.set_xlim(x.min() - pad, x.max() + pad)
        axt.set_ylim(y.min() - pad, y.max() + pad)
        axt.set_zlim(z.min() - pad, z.max() + pad)
        axt.set_xlabel(f'x [{up}]')
        axt.set_ylabel(f'y [{up}]')
        axt.set_zlabel(f'z [{up}]')
        axt.legend(loc='upper left', fontsize=8)
        # Escolhe um ângulo de vista que evite mostrar o plano orbital "de
        # perfil" (o que faria uma órbita elíptica parecer um segmento de
        # reta) — mistura a direção do momento angular com uma inclinação
        # padrão, para uma vista sempre razoavelmente oblíqua.
        if Lmag[0] > 1e-9:
            Lhat_view = np.array([Lx[0], Ly[0], Lz[0]]) / Lmag[0]
            theta_L = math.degrees(math.acos(np.clip(Lhat_view[2], -1, 1)))
            phi_L = math.degrees(math.atan2(Lhat_view[1], Lhat_view[0]))
            elev_v = 0.5 * (90 - theta_L) + 17.5
            azim_v = phi_L + 40
            axt.view_init(elev=elev_v, azim=azim_v)
        cbar = fig_traj.colorbar(lc, ax=axt, shrink=0.6, pad=0.1)
        cbar.set_label(f'|v|  [{up}/{ut}]')

    axt.set_title('B) Trajetória colorida pela velocidade escalar |v(t)|',
                   fontsize=12, fontweight='bold', color='#0d47a1')
    fig_traj.tight_layout()
    state['fig_traj'] = fig_traj
    with out_traj:
        display(HTML(fig_para_html(fig_traj, max_width=820)))
    plt.close(fig_traj)

    # --- Figura C: grade (posição, velocidade, aceleração) ---
    fig_grid, axs = plt.subplots(3, dim, figsize=(4.6 * dim, 9.5), sharex=True, squeeze=False)
    dados_q = [q1, q2] + ([q3] if coord_key == 'sph' else [])
    dados_v = [v1, v2] + ([v3] if coord_key == 'sph' else [])
    dados_a = [a1, a2] + ([a3] if coord_key == 'sph' else [])

    for j in range(dim):
        unit_q = up if cfg['q_kind'][j] == 'pos' else 'rad'
        axs[0, j].plot(ts, dados_q[j], color=COR_POS, lw=1.8)
        axs[0, j].set_ylabel(f"{cfg['q_disp'][j]}(t)  [{unit_q}]")
        axs[0, j].set_title(f"C) {cfg['q_disp'][j]}(t)", fontsize=10, loc='left', color=COR_POS)

        axs[1, j].plot(ts, dados_v[j], color=COR_VEL, lw=1.8)
        axs[1, j].set_ylabel(f"{cfg['v_disp'][j]}(t)  [{up}/{ut}]")
        axs[1, j].set_title(f"C) {cfg['v_disp'][j]}(t)", fontsize=10, loc='left', color=COR_VEL)

        axs[2, j].plot(ts, dados_a[j], color=COR_ACE, lw=1.8)
        axs[2, j].set_ylabel(f"{cfg['a_disp'][j]}(t)  [{up}/{ut}²]")
        axs[2, j].set_title(f"C) {cfg['a_disp'][j]}(t)", fontsize=10, loc='left', color=COR_ACE)
        axs[2, j].set_xlabel(f't  [{ut}]')

    for ax in axs.flat:
        ax.grid(True, alpha=0.3)

    fig_grid.suptitle('C) Posição, velocidade e aceleração (componentes físicas) em função do tempo',
                       fontsize=13, fontweight='bold', color='#0d47a1')
    fig_grid.tight_layout(rect=[0, 0, 1, 0.95])
    state['fig_grid'] = fig_grid
    with out_grid:
        display(HTML(fig_para_html(fig_grid, max_width=950 if dim == 2 else 1150)))
    plt.close(fig_grid)

    # --- Figura D (extra): órbita em coordenadas polares, no plano do movimento ---
    if coord_key == 'polar':
        angulo_plano = q2
        r_plano = q1
    else:
        Lhat = np.array([Lx[0], Ly[0], Lz[0]])
        Lhat = Lhat / np.linalg.norm(Lhat)
        r0vec = np.array([x[0], y[0], z[0]])
        e1 = r0vec - np.dot(r0vec, Lhat) * Lhat
        e1 = e1 / np.linalg.norm(e1)
        e2 = np.cross(Lhat, e1)
        rvec = np.stack([x, y, z], axis=1)
        proj1 = rvec @ e1
        proj2 = rvec @ e2
        angulo_plano = np.arctan2(proj2, proj1)
        r_plano = q1  # já é a distância à origem em ambos os sistemas

    fig_orb = plt.figure(figsize=(6.8, 6.8))
    axo = fig_orb.add_subplot(111, projection='polar')
    pontos_o = np.array([angulo_plano, r_plano]).T.reshape(-1, 1, 2)
    segmentos_o = np.concatenate([pontos_o[:-1], pontos_o[1:]], axis=1)
    lc_o = LineCollection(segmentos_o, cmap='viridis',
                           norm=plt.Normalize(speed.min(), speed.max()))
    lc_o.set_array(speed[:-1])
    lc_o.set_linewidth(2.6)
    axo.add_collection(lc_o)
    axo.scatter([angulo_plano[0]], [r_plano[0]], color='#2f9e44', s=55, zorder=5, label='Início')
    axo.scatter([angulo_plano[-1]], [r_plano[-1]], color='#c92a2a', s=55, zorder=5, label='Fim')
    axo.set_rlim(0, r_plano.max() * 1.08)
    axo.set_title('D) Extra: órbita no plano do movimento\n(r em função do ângulo no plano, colorida por |v|)',
                   fontsize=11, fontweight='bold', color='#0d47a1', pad=22)
    axo.legend(loc='upper right', fontsize=8, bbox_to_anchor=(1.25, 1.1))
    cbar_o = fig_orb.colorbar(lc_o, ax=axo, shrink=0.7, pad=0.1)
    cbar_o.set_label(f'|v|  [{up}/{ut}]')
    fig_orb.tight_layout()
    state['fig_orb'] = fig_orb
    with out_orbita:
        display(HTML(fig_para_html(fig_orb, max_width=720)))
        texto_extra = (
            'ℹ Como a força é central, o momento angular garante que o movimento '
            'fica sempre confinado a um plano fixo que passa pela origem — este '
            'gráfico mostra a órbita "de frente", vista perpendicularmente a esse '
            'plano, tornando visível a forma da trajetória (elipse, espiral, '
            'precessão etc.) de um jeito que a vista 3D nem sempre deixa claro.'
        )
        display(HTML(card(texto_extra, 'icard')))
    plt.close(fig_orb)

    with out_export:
        display(HTML(card(
            '✔ Simulação concluída. Use os botões abaixo para exportar os resultados.',
            'gcard')))

# ----------------------- Exportação -----------------------

def _baixar(caminho):
    if IN_COLAB:
        colab_files.download(caminho)
    else:
        display(HTML(card(f'Arquivo salvo em: <code>{caminho}</code>', 'icard')))


def _monta_df_dados():
    cfg = COORD_DEFS[state['coord_name']]
    up, ut, dim = state['up'], state['ut'], state['dim']
    unit_q = [up if k == 'pos' else 'rad' for k in cfg['q_kind']]
    cols = {f"t [{ut}]": state['ts']}
    for i in range(dim):
        cols[f"{cfg['q_disp'][i]}(t) [{unit_q[i]}]"] = state[f'q{i+1}']
    for i in range(dim):
        cols[f"{cfg['v_disp'][i]}(t) [{up}/{ut}]"] = state[f'v{i+1}']
    for i in range(dim):
        cols[f"{cfg['a_disp'][i]}(t) [{up}/{ut}²]"] = state[f'a{i+1}']
    cols[f"x(t) [{up}]"] = state['x']
    cols[f"y(t) [{up}]"] = state['y']
    if dim == 3:
        cols[f"z(t) [{up}]"] = state['z']
    cols["K(t) [J, assumindo SI]"] = state['K']
    cols["U(t) [J, assumindo SI]"] = state['U']
    cols["E_mec(t) = K+U [J, assumindo SI]"] = state['Emec']
    cols["L_x(t)"] = state['Lx']
    cols["L_y(t)"] = state['Ly']
    cols["L_z(t)"] = state['Lz']
    cols["|L(t)|"] = state['Lmag']
    return pd.DataFrame(cols)


def _monta_df_parametros():
    cfg = COORD_DEFS[state['coord_name']]
    dim = state['dim']
    rows = [
        ('Sistema de coordenadas', cfg['titulo']),
        ('Massa m (kg)', state['massa']),
        ('Fornecido pelo usuário', 'U(r)' if state['fornecendo_U'] else 'F_r(r)'),
        (f"U({cfg['q_syms'][0]})", state['U_expr']),
        (f"F_r({cfg['q_syms'][0]})", state['Fr_expr']),
        ('Momento angular inicial L0', state['L0']),
    ]
    for i in range(dim):
        rows.append((f"{cfg['q_disp'][i]}₀", state['q0'][i]))
    for i in range(dim):
        rows.append((f"{cfg['v_disp'][i]}₀", state['v0'][i]))
    rows += [
        ('t mínimo', state['tmin']),
        ('t máximo', state['tmax']),
        ('Unidade de posição', state['up']),
        ('Unidade de tempo', state['ut']),
        ('Método do integrador', state['metodo']),
        ('rtol', state['rtol']),
        ('atol', state['atol']),
        ('Nº de pontos', state['npts']),
    ]
    return pd.DataFrame(rows, columns=['Parâmetro', 'Valor'])


def on_export_png(btn=None):
    with out_export:
        if 'fig_traj' not in state:
            display(HTML(card('⚠ Execute a simulação antes de exportar.', 'wcard')))
            return
        caminhos = []
        for nome, fig in (('energia_momento_angular', state['fig_en']),
                           ('trajetoria_colorida_velocidade', state['fig_traj']),
                           ('posicao_velocidade_aceleracao', state['fig_grid']),
                           ('orbita_plano_movimento', state['fig_orb'])):
            caminho = f'/content/forcas_centrais_{nome}.png'
            fig.savefig(caminho, dpi=150, bbox_inches='tight')
            caminhos.append(caminho)
        display(HTML(card('✔ Gráficos exportados (4 arquivos PNG).', 'gcard')))
        for c in caminhos:
            _baixar(c)


def on_export_xlsx(btn=None):
    with out_export:
        if 'ts' not in state:
            display(HTML(card('⚠ Execute a simulação antes de exportar.', 'wcard')))
            return
        caminho = '/content/forcas_centrais_resultados.xlsx'
        with pd.ExcelWriter(caminho) as writer:
            _monta_df_dados().to_excel(writer, sheet_name='Dados_t_q_v_a_K_U_L', index=False)
            _monta_df_parametros().to_excel(writer, sheet_name='Parametros', index=False)
        display(HTML(card(f'✔ Planilha exportada: <code>{caminho}</code>', 'gcard')))
        _baixar(caminho)


def on_export_csv(btn=None):
    with out_export:
        if 'ts' not in state:
            display(HTML(card('⚠ Execute a simulação antes de exportar.', 'wcard')))
            return
        caminho = '/content/forcas_centrais_resultados_csv.zip'
        with zipfile.ZipFile(caminho, 'w') as zf:
            zf.writestr('Dados_t_q_v_a_K_U_L.csv', _monta_df_dados().to_csv(index=False))
            zf.writestr('Parametros.csv', _monta_df_parametros().to_csv(index=False))
        display(HTML(card(f'✔ Dados exportados (ZIP com CSVs): <code>{caminho}</code>', 'gcard')))
        _baixar(caminho)


simular_btn.on_click(on_simular_clicked)
export_png_btn.on_click(on_export_png)
export_xlsx_btn.on_click(on_export_xlsx)
export_csv_btn.on_click(on_export_csv)


# ----------------------- Interface -----------------------

ui = w.VBox([
    cabecalho,

    titulo_secao('⚙️ Partícula, sistema de coordenadas e força central'),
    w.HBox([massa_text, coord_dropdown]),
    obs_coord_html,
    entrada_dropdown,
    entrada_text,

    titulo_secao('🎯 Condições iniciais e intervalo de simulação'),
    w.HBox([tmin_text, tmax_text]),
    w.HBox([q1_0, q2_0, linha_q3]),
    w.HBox([v1_0, v2_0, linha_v3]),
    w.HBox([unidade_pos, unidade_tempo]),
    unidades_obs,

    opcoes_avancadas,
    simular_btn,

    titulo_secao('📋 Informações da simulação'),
    out_info,

    titulo_secao('📊 A) Energia efetiva e momento angular'),
    out_energia,

    titulo_secao('📊 B) Trajetória colorida pela velocidade'),
    out_traj,

    titulo_secao('📊 C) Posição, velocidade e aceleração vs. tempo'),
    out_grid,

    titulo_secao('📊 D) Extra: órbita no plano do movimento'),
    out_orbita,

    titulo_secao('💾 Exportar resultados'),
    w.HBox([export_png_btn, export_xlsx_btn, export_csv_btn]),
    out_export,
])

display(ui)
```
