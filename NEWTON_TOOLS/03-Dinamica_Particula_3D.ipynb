Esta ferramenta simula a **dinâmica 3D de uma partícula** de massa $m$ sob a ação da força resultante $\vec{F}(t,\vec{r},\vec{v})$. A solução é obtida numericamente a partir do **problema de valor inicial (PVI)** associado à **2ª Lei de Newton**. O PVI é composto pelas equações diferenciais ordinárias (EDOs) e pelas condições iniciais do problema.

A ferramenta admite três **sistemas de coordenadas**:

- **Cartesianas**: $x$, $y$, $z$
- **Cilíndricas**: $\rho$, $\varphi$, $z$
- **Esféricas**: $r$, $\varphi$, $\theta$

Nas coordenadas cilíndricas e esféricas, $\varphi$ é o ângulo azimutal (medido no plano $xy$); nas coordenadas esféricas, $\theta$ é o ângulo polar (medido a partir do semieixo $z$ positivo), sempre em **radianos**.

As coordenadas cartesianas $x$, $y$ e $z$ são expressas, em termos das coordenadas cilíndricas e das coordenadas esféricas e cartesianas, por:

| Coordenadas cilíndricas | Coordenadas esféricas |
|:---|:---|
| $x=\rho\cos\varphi$   | $x=r\cos\varphi\sin\theta$   |
| $y=\rho\sin\varphi$ |  $y=r\sin\varphi\sin\theta$ |
| $z=z$ | $z=r\cos\theta$ |

A ferramenta aceita que as componentes da força sejam expressas em qualquer um destes sistemas: $F_x, F_y, F_z$, $F_\rho, F_\varphi, F_z$, ou $F_r, F_\varphi, F_\theta$. A força pode depender do tempo `t`, da massa `m`, das coordenadas da partícula e das componentes da velocidade.

É importante notar que as componentes das velocidade e da aceleração usadas em todos os campos, gráficos e na força digitada são as **componentes físicas** da velocidade (com dimensão $\mathrm{L}\mathrm{T}^{-1}$) ou da aceleração (com dimensão $\mathrm{L}\mathrm{T}^{-2}$) e não grandezas como as velocidades angulares $\dot{\varphi}$ e $\dot{\theta}$ ou as acelerações angulares $\ddot{\varphi}$ e $\ddot{\theta}$. ---
### Sintaxe da expressão da força

As variáveis disponíveis são `t` (tempo), `m` (massa) e as coordenadas/velocidades do sistema escolhido (por exemplo `x, y, z, vx, vy, vz` para Cartesianas, `rho, phi, z, vrho, vphi, vz` para Cilíndricas, ou `r, phi, theta, vr, vphi, vtheta` para Esféricas). As expressões devem ser digitadas usando a notação do pacote SymPy (https://www.sympy.org/):

| Para obter &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; | Digite &nbsp; &nbsp; &nbsp; &nbsp; |
|:---|:---|
| $-5.7\,x$ | `-5.7*x` |
| $-m\,g$ | `-m*9.8` |
| $\sqrt{r/2}$ | `sqrt(r/2)` |
| $e^{-2t}$ | `exp(-2*t)` |
| $\sin(2\pi t)$ | `sin(2*pi*t)` |
| $-k\,\rho - b\,v_\rho$ | `-2*rho - 0.3*vrho` |
| $-G M_\odot m / r^2$ | `-1.327e20*m/r**2` |

1. **Constantes** (como $k$, $b$, $g$, $GM_\odot$) devem ser digitadas como **números**, não como símbolos, porque a ferramenta não reconhece letras que não sejam `t`, `m` ou as coordenadas/velocidades do sistema escolhido.
2. **Evite usar o símbolo `I`**, porque `I` é reservado pelo SymPy para representar a unidade imaginária $i=\sqrt{-1}$.
3. **Evite usar o símbolo `E`**, porque `E` é reservado pelo SymPy para representar a base do logaritmo natural: `E`$=\exp(1)$.
4. **Potências de 10** como $1.327\times 10^{20}$ podem ser expressas como `1.327e20` (`e` obrigatoriamente **minúsculo**) ou como `1.327*10**20`.
5. Além disso, os símbolos `S`, `N`, `C`, `O`, `Q`, `zoo`, `nan`, `oo`, `true`, `false`, `True` e `False` não devem ser usados, porque também são símbolos reservados pelo SymPy.

---
### O que a ferramenta calcula

Você informa a massa $m$, o sistema de coordenadas, as três componentes da força, o intervalo de simulação $[t_\mathrm{min}, t_\mathrm{max}]$ e as condições iniciais (posição e velocidade, sempre no instante inicial $t=t_\mathrm{min}$). A ferramenta resolve numericamente o sistema de equações diferenciais da 2ª Lei de Newton usando `scipy.integrate.solve_ivp` (você pode escolher o método do integrador e ajustar tolerâncias nas opções avançadas) e mostra:

- **a)** Trajetória da partícula em um gráfico 3D em coordenadas cartesianas;
- **b, c, d)** as três coordenadas de posição em função de $t$;
- **e, f, g)** as três componentes (físicas) da velocidade em função de $t$;
- **h, i, j)** as três componentes (físicas) da aceleração em função de $t$;
- **k)** a energia cinética $K(t)=\tfrac12 m v^2$;
- **l)** o momento angular da partícula em relação à origem, $|\vec L(t)|$ e $L_z(t)$ — útil para checar visualmente a conservação do momento angular (por exemplo, em forças centrais) e a qualidade da integração numérica.

---
**Atenção**: os campos de unidade de posição e de tempo servem apenas para rotular eixos e tabelas e **não realizam conversão entre unidades**. Cabe a você garantir que $m$, $\vec F$, as posições e o tempo estejam todos em um sistema de unidades coerente (recomendado: SI — kg, m, s, N). Os ângulos $\varphi$ e $\theta$ são sempre tratados em radianos.

**Atenção**: a origem ($\rho=0$ ou $r=0$) é uma singularidade dos sistemas cilíndrico e esférico (o ângulo azimutal não é definido ali) — evite condições iniciais ou trajetórias que passem exatamente pela origem nesses sistemas. Da mesma forma, os polos ($\theta=0$ ou $\theta=\pi$) são singulares no sistema esférico.

```python
#@title Execute esta célula para abrir a ferramenta **NEWTON TOOLS &mdash; Dinâmica da Partícula 3D**
# ── Célula 1: Dinâmica 3D (Problema de Valor Inicial) ───────────────────────
# A ferramenta foi especificada pelo licenciante, incluindo modelos, métodos,
# pacotes a serem usados e linhas gerais da inteface gráfica, mas a maior parte
# do código, especialmente as funções para leitura de arquivos e para a
# interface gráfica, foi escrita usando o modelo Claude Sonnet 5.0.
# O código gerado foi inspecionado e sofreu alterações realizadas pelo licenciante.

import math, io, base64, warnings
import numpy as np
import sympy as sp
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sympy.parsing.sympy_parser import (
    parse_expr, standard_transformations, implicit_multiplication_application
)
from scipy.integrate import solve_ivp
import ipywidgets as w
from IPython.display import display, HTML, Math, clear_output
import zipfile

warnings.filterwarnings("ignore")
matplotlib.rcParams.update({"figure.dpi": 110, "font.size": 10,
                             "axes.spines.top": False,
                             "axes.spines.right": False})

# Detecta se está rodando no Google Colab
try:
    from google.colab import files as colab_files
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# Transformações para aceitar entradas como "2*m" com multiplicação implícita
TRANSFORMATIONS = standard_transformations + (implicit_multiplication_application,)

# ══════════════════════════════════════════════════════════════════
#  ESTILOS CSS
#  Convenção compartilhada por toda a suíte LAB/NEWTON TOOLS:
#    .sec   → rótulo de seção (cor = acento do app)
#    .rcard → card de resultado/destaque (cor = acento do app)
#    .gcard → card de "sucesso" (verde fixo)
#    .bcard → card informativo secundário (azul fixo)
#    .wcard → card de aviso (laranja fixo)
#    .ecard → card de erro (vermelho fixo)
#    .icard → card informativo auxiliar (violeta fixo)
#    .res-t → tabela de dados (cabeçalho azul-marinho fixo)
#  Acento desta ferramenta: ciano #1098ad
# ══════════════════════════════════════════════════════════════════
display(HTML("""
<style>
.sec{font-weight:700;font-size:.75rem;letter-spacing:.08em;color:#1098ad;
  text-transform:uppercase;margin:13px 0 3px;}
.rcard{background:#eafcff;border-left:4px solid #1098ad;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-family:monospace;font-size:.91rem;}
.gcard{background:#f0fff4;border-left:4px solid #2f9e44;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.91rem;}
.bcard{background:#f0f4ff;border-left:4px solid #3b5bdb;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-family:monospace;font-size:.91rem;}
.wcard{background:#fff9db;border-left:4px solid #e67700;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.91rem;}
.ecard{background:#fff0f0;border-left:4px solid #c92a2a;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.91rem;}
.icard{background:#f3f0ff;border-left:4px solid #7048e8;border-radius:6px;
  padding:9px 15px;margin:5px 0;font-size:.88rem;}
.res-t{border-collapse:collapse;font-size:.83rem;width:60%;margin:6px 0;}
.res-t th{background:#1a1a2e;color:#fff;padding:5px 10px;text-align:left;}
.res-t td{padding:4px 10px;text-align:left;border-bottom:1px solid #eee;
  font-family:monospace;}
.res-t tr:nth-child(odd) td{text-align:left}
.res-t tr:nth-child(even) td{background:#eafcff;text-align:left}
</style>
"""))

# ══════════════════════════════════════════════════════════════════
#  HTML HELPERS
# ══════════════════════════════════════════════════════════════════
def h(tag, content, **attrs):
    a = " ".join(f'{k.rstrip("_")}="{v}"' for k, v in attrs.items())
    return f"<{tag} {a}>{content}</{tag}>"

def card(s, c="rcard"): return f'<div class="{c}">{s}</div>'

def sec_html(texto, estilo=""):
    """Constrói o HTML de um rótulo de seção (classe .sec)."""
    extra = f' style="{estilo}"' if estilo else ""
    return f'<div class="sec"{extra}>{texto}</div>'

def titulo_secao(texto, estilo=""):
    """Widget HTML com um rótulo de seção — forma padrão de criar
    títulos de seção em toda a suíte LAB/NEWTON TOOLS."""
    return w.HTML(sec_html(texto, estilo))

def tabela_html(df):
    """Renderiza um DataFrame com o mesmo estilo .res-t usado nas
    demais ferramentas da suíte."""
    return df.to_html(classes="res-t", index=False, border=0,
                       float_format=lambda v: f"{v:.6g}")

def fig_para_html(fig, max_width=900):
    """Converte uma figura matplotlib em <img> embutido (PNG base64),
    no mesmo padrão visual usado nas demais ferramentas da suíte.
    Não fecha a figura — ela é reutilizada na exportação em PNG."""
    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=130, bbox_inches="tight")
    buf.seek(0)
    b64 = base64.b64encode(buf.read()).decode()
    return (f'<img src="data:image/png;base64,{b64}" '
            f'style="width:100%;max-width:{max_width}px;border-radius:6px;'
            f'box-shadow:0 2px 8px rgba(0,0,0,.13)">')

# ══════════════════════════════════════════════════════════════════
#  MOTOR FÍSICO — sistemas de coordenadas, símbolos e equações de movimento
#
# ══════════════════════════════════════════════════════════════════

t_sym = sp.symbols('t', real=True)
m_sym = sp.symbols('m', positive=True)

COORD_DEFS = {
    'Cartesianas': dict(
        key='cart',
        titulo='Cartesianas (x, y, z)',
        q_syms=['x', 'y', 'z'], v_syms=['vx', 'vy', 'vz'],
        q_disp=['x', 'y', 'z'], v_disp=['v_x', 'v_y', 'v_z'],
        F_disp=['F_x', 'F_y', 'F_z'], a_disp=['a_x', 'a_y', 'a_z'],
        q_kind=['pos', 'pos', 'pos'],
        q_default=[0.0, 0.0, 0.0], v_default=[5.0, 3.0, 8.0],
        F_default=['0', '0', '-m*9.8'],
        obs=('Sistema de coordenadas cartesianas: x, y e z são as '
             'coordenadas usuais e v_x, v_y, v_z as componentes usuais '
             'da velocidade.'),
    ),
    'Cilíndricas': dict(
        key='cyl',
        titulo='Cilíndricas (ρ, φ, z)',
        q_syms=['rho', 'phi', 'z'], v_syms=['vrho', 'vphi', 'vz'],
        q_disp=['ρ', 'φ', 'z'], v_disp=['v_ρ', 'v_φ', 'v_z'],
        F_disp=['F_ρ', 'F_φ', 'F_z'], a_disp=['a_ρ', 'a_φ', 'a_z'],
        q_kind=['pos', 'ang', 'pos'],
        q_default=[1.0, 0.0, 0.0], v_default=[0.0, 1.5, 0.0],
        F_default=['-2*rho', '0', '0'],
        obs=('ρ é a distância ao eixo z e φ é o ângulo azimutal (medido no '
             'plano xy, em radianos). v_ρ = dρ/dt e v_φ = ρ·dφ/dt é a '
             'componente FÍSICA (tangencial) da velocidade — não confundir '
             'com dφ/dt.'),
    ),
    'Esféricas': dict(
        key='sph',
        titulo='Esféricas (r, φ, θ)',
        q_syms=['r', 'phi', 'theta'], v_syms=['vr', 'vphi', 'vtheta'],
        q_disp=['r', 'φ', 'θ'], v_disp=['v_r', 'v_φ', 'v_θ'],
        F_disp=['F_r', 'F_φ', 'F_θ'], a_disp=['a_r', 'a_φ', 'a_θ'],
        q_kind=['pos', 'ang', 'ang'],
        q_default=[2.0, 0.0, 1.5707963267948966],  # theta0 = pi/2 (plano equatorial)
        v_default=[0.0, 1.0, 0.0],
        F_default=['-4*m/r**2', '0', '0'],
        obs=('r é a distância à origem, φ é o ângulo azimutal (plano xy) e '
             'θ é o ângulo polar medido a partir do semieixo z positivo '
             '(ambos em radianos). v_φ = r·sin(θ)·dφ/dt e v_θ = r·dθ/dt são '
             'as componentes FÍSICAS (tangenciais) da velocidade.'),
    ),
}

RESERVED = {'I', 'E', 'S', 'N', 'C', 'O', 'Q', 'zoo', 'nan', 'oo',
            'true', 'false', 'True', 'False'}


def local_dict_for(coord_key_name):
    """Monta o dicionário de símbolos SymPy (t, m, q1,q2,q3, v1,v2,v3)
    para o sistema de coordenadas escolhido."""
    cfg = COORD_DEFS[coord_key_name]
    qs = [sp.symbols(s, real=True) for s in cfg['q_syms']]
    vs = [sp.symbols(s, real=True) for s in cfg['v_syms']]
    ld = {'t': t_sym, 'm': m_sym}
    for name, sym in zip(cfg['q_syms'], qs):
        ld[name] = sym
    for name, sym in zip(cfg['v_syms'], vs):
        ld[name] = sym
    return ld, qs, vs


def parse_forca(texto, local_dict, allowed_names):
    expr = parse_expr(texto, local_dict=local_dict, transformations=TRANSFORMATIONS)
    usados = {str(s) for s in expr.free_symbols}
    bad_reserved = usados & RESERVED
    if bad_reserved:
        raise ValueError(f"símbolo(s) reservado(s) do SymPy usados: {', '.join(sorted(bad_reserved))}")
    bad_unknown = usados - allowed_names
    if bad_unknown:
        raise ValueError(
            f"símbolo(s) não reconhecido(s): {', '.join(sorted(bad_unknown))}. "
            f"Use apenas t, m e as coordenadas/velocidades do sistema escolhido "
            f"({', '.join(sorted(allowed_names - {'t','m'}))}); constantes devem "
            f"ser digitadas como números.")
    return expr


def make_force_funcs(expr, arg_syms):
    """Retorna (f_scalar, f_vec): funções numéricas que avaliam a força
    simbólica em (t, m, q1, q2, q3, v1, v2, v3), tanto para argumentos
    escalares (usados dentro do integrador) quanto vetoriais/numpy
    (usados no pós-processamento para os gráficos)."""
    if expr.free_symbols == set():
        val = float(expr)
        def f_scalar(*a):
            return val
        def f_vec(*a):
            ref = np.asarray(a[0], dtype=float)
            return np.full_like(ref, val)
        return f_scalar, f_vec

    f_np = sp.lambdify(arg_syms, expr, 'numpy')

    def f_scalar(*a):
        return float(f_np(*a))

    def f_vec(*a):
        y = np.asarray(f_np(*a), dtype=float)
        ref = np.asarray(a[0], dtype=float)
        if y.shape != ref.shape:
            y = np.full_like(ref, float(y))
        return y

    return f_scalar, f_vec


# ---------------------------------------------------------------------
# Equações de movimento (2ª Lei de Newton) para cada sistema de coordenadas.
# Estado y = [q1, q2, q3, v1, v2, v3].
# ---------------------------------------------------------------------

EPS = 1e-9

def make_rhs(coord_key, massa, F_scalars):
    F1, F2, F3 = F_scalars

    if coord_key == 'cart':
        def rhs(tt, y):
            x, yy, z, vx, vy, vz = y
            Fx = F1(tt, massa, x, yy, z, vx, vy, vz)
            Fy = F2(tt, massa, x, yy, z, vx, vy, vz)
            Fz = F3(tt, massa, x, yy, z, vx, vy, vz)
            return [vx, vy, vz, Fx / massa, Fy / massa, Fz / massa]
        return rhs

    if coord_key == 'cyl':
        def rhs(tt, y):
            rho, phi, z, vrho, vphi, vz = y
            rho_s = rho if abs(rho) > EPS else (EPS if rho >= 0 else -EPS)
            Frho = F1(tt, massa, rho, phi, z, vrho, vphi, vz)
            Fphi = F2(tt, massa, rho, phi, z, vrho, vphi, vz)
            Fz_ = F3(tt, massa, rho, phi, z, vrho, vphi, vz)
            drho = vrho
            dphi = vphi / rho_s
            dz = vz
            dvrho = Frho / massa + vphi**2 / rho_s
            dvphi = Fphi / massa - vrho * vphi / rho_s
            dvz = Fz_ / massa
            return [drho, dphi, dz, dvrho, dvphi, dvz]
        return rhs

    if coord_key == 'sph':
        def rhs(tt, y):
            r, phi, theta, vr, vphi, vtheta = y
            r_s = r if abs(r) > EPS else (EPS if r >= 0 else -EPS)
            sin_t = math.sin(theta)
            sin_s = sin_t if abs(sin_t) > EPS else (EPS if sin_t >= 0 else -EPS)
            cot_t = math.cos(theta) / sin_s
            Fr = F1(tt, massa, r, phi, theta, vr, vphi, vtheta)
            Fphi = F2(tt, massa, r, phi, theta, vr, vphi, vtheta)
            Ftheta = F3(tt, massa, r, phi, theta, vr, vphi, vtheta)
            dr = vr
            dphi = vphi / (r_s * sin_s)
            dtheta = vtheta / r_s
            dvr = Fr / massa + (vtheta**2 + vphi**2) / r_s
            dvphi = Fphi / massa - vr * vphi / r_s - vtheta * vphi * cot_t / r_s
            dvtheta = Ftheta / massa - vr * vtheta / r_s + vphi**2 * cot_t / r_s
            return [dr, dphi, dtheta, dvr, dvphi, dvtheta]
        return rhs

    raise ValueError(coord_key)


def calc_aceleracoes(coord_key, massa, F_vecs, ts, q1, q2, q3, v1, v2, v3):
    """Recalcula a1, a2, a3 = dv/dt (vetorizado) a partir da solução,
    usando exatamente as mesmas equações do integrador."""
    F1, F2, F3 = F_vecs
    F1v = F1(ts, massa, q1, q2, q3, v1, v2, v3)
    F2v = F2(ts, massa, q1, q2, q3, v1, v2, v3)
    F3v = F3(ts, massa, q1, q2, q3, v1, v2, v3)

    if coord_key == 'cart':
        a1 = F1v / massa
        a2 = F2v / massa
        a3 = F3v / massa
    elif coord_key == 'cyl':
        rho_s = np.where(np.abs(q1) > EPS, q1, EPS)
        a1 = F1v / massa + v2**2 / rho_s
        a2 = F2v / massa - v1 * v2 / rho_s
        a3 = F3v / massa
    elif coord_key == 'sph':
        r_s = np.where(np.abs(q1) > EPS, q1, EPS)
        sin_t = np.sin(q3)
        sin_s = np.where(np.abs(sin_t) > EPS, sin_t, EPS)
        cot_t = np.cos(q3) / sin_s
        a1 = F1v / massa + (v3**2 + v2**2) / r_s
        a2 = F2v / massa - v1 * v2 / r_s - v3 * v2 * cot_t / r_s
        a3 = F3v / massa - v1 * v3 / r_s + v2**2 * cot_t / r_s
    else:
        raise ValueError(coord_key)
    return a1, a2, a3


def para_cartesianas(coord_key, q1, q2, q3, v1, v2, v3):
    """Converte posição e velocidade para coordenadas cartesianas
    (x, y, z, vx, vy, vz), para uso na trajetória 3D, energia e
    momento angular."""
    if coord_key == 'cart':
        return q1, q2, q3, v1, v2, v3

    if coord_key == 'cyl':
        rho, phi, z = q1, q2, q3
        vrho, vphi, vz = v1, v2, v3
        x = rho * np.cos(phi)
        y = rho * np.sin(phi)
        vx = vrho * np.cos(phi) - vphi * np.sin(phi)
        vy = vrho * np.sin(phi) + vphi * np.cos(phi)
        return x, y, z, vx, vy, vz

    if coord_key == 'sph':
        r, phi, theta = q1, q2, q3
        vr, vphi, vtheta = v1, v2, v3
        sin_t, cos_t = np.sin(theta), np.cos(theta)
        sin_p, cos_p = np.sin(phi), np.cos(phi)
        x = r * sin_t * cos_p
        y = r * sin_t * sin_p
        z = r * cos_t
        vx = vr * sin_t * cos_p + vtheta * cos_t * cos_p - vphi * sin_p
        vy = vr * sin_t * sin_p + vtheta * cos_t * sin_p + vphi * cos_p
        vz = vr * cos_t - vtheta * sin_t
        return x, y, z, vx, vy, vz

    raise ValueError(coord_key)

# ----------------------- Cabeçalho -----------------------

cabecalho = w.HTML(r'''
<div style="background: linear-gradient(135deg, #004587 0%, #00AB45 80%, #006EAB 100%);
            padding: 14px 20px; border-radius: 10px; color: white;
            font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 12px;
            box-shadow: 0 2px 6px rgba(0,0,0,0.15);">
  <div style="font-size: 13px; letter-spacing: 3px; opacity: 0.85; font-weight: 600;">QapybaLAB &mdash; NEWTON TOOLS</div>
  <div style="font-size: 26px; font-weight: 700; margin-top: 2px;">🚀 Dinâmica da Partícula 3D</div>
</div>
''')

# ----------------------- Widgets de entrada -----------------------

massa_text = w.FloatText(
    value=1.0, description='Massa m (kg):',
    style={'description_width': 'initial'}, layout=w.Layout(width='220px')
)

coord_dropdown = w.Dropdown(
    options=list(COORD_DEFS.keys()), value='Cartesianas',
    description='Sistema de coordenadas:',
    style={'description_width': 'initial'}, layout=w.Layout(width='320px')
)

obs_coord_html = w.HTML()

F1_text = w.Text(layout=w.Layout(width='420px'), style={'description_width': '55px'})
F2_text = w.Text(layout=w.Layout(width='420px'), style={'description_width': '55px'})
F3_text = w.Text(layout=w.Layout(width='420px'), style={'description_width': '55px'})

tmin_text = w.FloatText(value=0.0, description='t mínimo:',
                         style={'description_width': 'initial'}, layout=w.Layout(width='170px'))
tmax_text = w.FloatText(value=5.0, description='t máximo:',
                         style={'description_width': 'initial'}, layout=w.Layout(width='170px'))

q1_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
q2_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
q3_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
v1_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
v2_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})
v3_0 = w.FloatText(layout=w.Layout(width='220px'), style={'description_width': '55px'})

unidade_pos = w.Text(value='m', description='Unidade de posição:',
                      style={'description_width': 'initial'}, layout=w.Layout(width='200px'))
unidade_tempo = w.Text(value='s', description='Unidade de tempo:',
                        style={'description_width': 'initial'}, layout=w.Layout(width='200px'))

unidades_obs = w.HTML(card(
    '⚠ Os campos de unidade servem apenas para rotular eixos e tabelas — '
    'não convertem valores. Garanta que m, F, posições e tempo estejam '
    'todos em um sistema de unidades coerente (recomendado: SI &mdash; kg, m, s, N).',
    'wcard'))

# --- opções numéricas avançadas ---
metodo_dropdown = w.Dropdown(
    options=['RK45', 'RK23', 'DOP853', 'Radau', 'BDF', 'LSODA'],
    value='RK45', description='Método do integrador:',
    style={'description_width': 'initial'}, layout=w.Layout(width='260px')
)
npts_text = w.IntText(value=600, description='Nº de pontos para os gráficos:',
                       style={'description_width': 'initial'}, layout=w.Layout(width='260px'))
rtol_text = w.FloatText(value=1e-8, description='Tolerância relativa (rtol):',
                         style={'description_width': 'initial'}, layout=w.Layout(width='260px'))
atol_text = w.FloatText(value=1e-10, description='Tolerância absoluta (atol):',
                         style={'description_width': 'initial'}, layout=w.Layout(width='260px'))

opcoes_avancadas = w.Accordion(children=[w.VBox([
    w.HBox([metodo_dropdown, npts_text]),
    w.HBox([rtol_text, atol_text]),
])])
opcoes_avancadas.set_title(0, '⚙️ Opções numéricas (avançado)')
opcoes_avancadas.selected_index = None  # inicia recolhido

simular_btn = w.Button(description='▶  Simular', button_style='primary', icon='rocket',
                        layout=w.Layout(width='160px', height='36px'))

export_png_btn = w.Button(description='Gráficos (PNG)', icon='image')
export_xlsx_btn = w.Button(description='Exportar XLSX', icon='file-excel')
export_csv_btn = w.Button(description='Exportar CSV', icon='file-text')

out_info = w.Output()
out_eqs = w.Output()
out_traj = w.Output()
out_grid = w.Output()
out_extra = w.Output()
out_export = w.Output()

# Equações do movimento (2ª Lei de Newton) em cada sistema de coordenadas,
# escritas em HTML puro para tornar renderização mais estável no Colab).
EQS_HTML = {
    'cart': (
        "dx/dt = v<sub>x</sub> &nbsp; dy/dt = v<sub>y</sub> &nbsp; dz/dt = v<sub>z</sub><br>"
        "m·dv<sub>x</sub>/dt = F<sub>x</sub> &nbsp; "
        "m·dv<sub>y</sub>/dt = F<sub>y</sub> &nbsp; "
        "m·dv<sub>z</sub>/dt = F<sub>z</sub>"
    ),
    'cyl': (
        "dρ/dt = v<sub>ρ</sub> &nbsp; dφ/dt = v<sub>φ</sub>/ρ &nbsp; dz/dt = v<sub>z</sub><br>"
        "m·dv<sub>ρ</sub>/dt = F<sub>ρ</sub> + m·v<sub>φ</sub>²/ρ<br>"
        "m·dv<sub>φ</sub>/dt = F<sub>φ</sub> − m·v<sub>ρ</sub>·v<sub>φ</sub>/ρ<br>"
        "m·dv<sub>z</sub>/dt = F<sub>z</sub>"
    ),
    'sph': (
        "dr/dt = v<sub>r</sub> &nbsp; dφ/dt = v<sub>φ</sub>/(r·senθ) &nbsp; dθ/dt = v<sub>θ</sub>/r<br>"
        "m·dv<sub>r</sub>/dt = F<sub>r</sub> + m·(v<sub>θ</sub>² + v<sub>φ</sub>²)/r<br>"
        "m·dv<sub>φ</sub>/dt = F<sub>φ</sub> − m·v<sub>r</sub>·v<sub>φ</sub>/r − m·v<sub>θ</sub>·v<sub>φ</sub>·cotθ/r<br>"
        "m·dv<sub>θ</sub>/dt = F<sub>θ</sub> − m·v<sub>r</sub>·v<sub>θ</sub>/r + m·v<sub>φ</sub>²·cotθ/r"
    ),
}


def atualizar_coordenadas(change=None):
    cfg = COORD_DEFS[coord_dropdown.value]

    F1_text.description = f"{cfg['F_disp'][0]} ="
    F2_text.description = f"{cfg['F_disp'][1]} ="
    F3_text.description = f"{cfg['F_disp'][2]} ="
    F1_text.value = cfg['F_default'][0]
    F2_text.value = cfg['F_default'][1]
    F3_text.value = cfg['F_default'][2]

    q1_0.description = f"{cfg['q_disp'][0]}₀ ="
    q2_0.description = f"{cfg['q_disp'][1]}₀ ="
    q3_0.description = f"{cfg['q_disp'][2]}₀ ="
    v1_0.description = f"{cfg['v_disp'][0]}₀ ="
    v2_0.description = f"{cfg['v_disp'][1]}₀ ="
    v3_0.description = f"{cfg['v_disp'][2]}₀ ="
    q1_0.value, q2_0.value, q3_0.value = cfg['q_default']
    v1_0.value, v2_0.value, v3_0.value = cfg['v_default']

    unid_ang_txt = ' (φ, θ sempre em radianos)' if 'ang' in cfg['q_kind'] else ''
    obs_coord_html.value = card(cfg['obs'] + unid_ang_txt, 'icard')

    out_eqs.clear_output()
    with out_eqs:
        display(HTML(card(EQS_HTML[cfg['key']], 'bcard')))


coord_dropdown.observe(atualizar_coordenadas, names='value')
atualizar_coordenadas()

# ----------------------- Lógica principal -----------------------

# Dicionário usado para guardar os resultados da última simulação (exportação)
state = {}

COR_POS = '#1565c0'
COR_VEL = '#ef6c00'
COR_ACE = '#2e7d32'
COR_TRAJ = '#1098ad'


def on_simular_clicked(btn=None):
    for o in (out_info, out_traj, out_grid, out_extra, out_export):
        o.clear_output()

    massa = massa_text.value
    if not np.isfinite(massa) or massa <= 0:
        with out_info:
            display(HTML(card('✘ A massa m deve ser um número positivo.', 'ecard')))
        return

    tmin, tmax = tmin_text.value, tmax_text.value
    if tmin >= tmax:
        with out_info:
            display(HTML(card('✘ O tempo mínimo deve ser menor que o tempo máximo.', 'ecard')))
        return

    npts = max(int(npts_text.value), 20)
    rtol, atol = rtol_text.value, atol_text.value
    metodo = metodo_dropdown.value

    coord_name = coord_dropdown.value
    cfg = COORD_DEFS[coord_name]
    coord_key = cfg['key']
    up = unidade_pos.value.strip() or 'm'
    ut = unidade_tempo.value.strip() or 's'

    ld, qs, vs = local_dict_for(coord_name)
    allowed_names = {'t', 'm'} | set(cfg['q_syms']) | set(cfg['v_syms'])
    arg_syms = (t_sym, m_sym, qs[0], qs[1], qs[2], vs[0], vs[1], vs[2])

    # --- Interpreta as três componentes da força ---
    try:
        F_exprs = [
            parse_forca(F1_text.value, ld, allowed_names),
            parse_forca(F2_text.value, ld, allowed_names),
            parse_forca(F3_text.value, ld, allowed_names),
        ]
    except Exception as e:
        with out_info:
            display(HTML(card(f'✘ Erro ao interpretar a força: {e}', 'ecard')))
        return

    F_scalars, F_vecs = [], []
    for expr in F_exprs:
        fs, fv = make_force_funcs(expr, arg_syms)
        F_scalars.append(fs)
        F_vecs.append(fv)

    # --- Condições iniciais ---
    q0 = [q1_0.value, q2_0.value, q3_0.value]
    v0 = [v1_0.value, v2_0.value, v3_0.value]

    avisos = []

    if coord_key in ('cyl', 'sph') and abs(q0[0]) < 1e-6:
        nome0 = cfg['q_disp'][0]
        with out_info:
            display(HTML(card(
                f'✘ A origem ({nome0} = 0) é uma singularidade do sistema de '
                f'coordenadas {cfg["titulo"]} (o ângulo azimutal não é definido '
                f'ali). Escolha {nome0}₀ > 0, ou use coordenadas cartesianas.',
                'ecard')))
        return

    if coord_key == 'sph':
        theta0 = q0[2]
        if abs(math.sin(theta0)) < 1e-6:
            avisos.append(
                'A condição inicial está muito próxima de um polo (θ₀ ≈ 0 ou π), '
                'onde o ângulo azimutal φ é indefinido. Os resultados podem ficar '
                'numericamente instáveis nessa região.')

    # --- Integração numérica (2ª Lei de Newton) ---
    rhs = make_rhs(coord_key, massa, F_scalars)
    t_eval = np.linspace(tmin, tmax, npts)
    try:
        sol = solve_ivp(rhs, (tmin, tmax), q0 + v0, t_eval=t_eval,
                         method=metodo, rtol=rtol, atol=atol)
    except Exception as e:
        with out_info:
            display(HTML(card(f'✘ Erro durante a integração numérica: {e}', 'ecard')))
        return

    if not sol.success:
        with out_info:
            display(HTML(card(
                f'✘ O integrador não convergiu: {sol.message}', 'ecard')))
        return

    ts = sol.t
    q1, q2, q3, v1, v2, v3 = sol.y
    a1, a2, a3 = calc_aceleracoes(coord_key, massa, F_vecs, ts, q1, q2, q3, v1, v2, v3)

    if coord_key in ('cyl', 'sph') and np.min(np.abs(q1)) < 1e-4 * max(abs(q0[0]), 1.0):
        avisos.append(
            f'A trajetória passou muito perto da origem ({cfg["q_disp"][0]} ≈ 0) '
            'durante a simulação — região singular do sistema de coordenadas. '
            'Verifique os resultados com cautela.')

    x, y, z, vx, vy, vz = para_cartesianas(coord_key, q1, q2, q3, v1, v2, v3)

    K = 0.5 * massa * (v1**2 + v2**2 + v3**2)
    Lx = massa * (y * vz - z * vy)
    Ly = massa * (z * vx - x * vz)
    Lz = massa * (x * vy - y * vx)
    Lmag = np.sqrt(Lx**2 + Ly**2 + Lz**2)

    state.clear()
    state.update(
        massa=massa, coord_name=coord_name, coord_key=coord_key,
        F_exprs=[str(e) for e in F_exprs], q0=q0, v0=v0,
        tmin=tmin, tmax=tmax, metodo=metodo, rtol=rtol, atol=atol, npts=npts,
        up=up, ut=ut, ts=ts, q1=q1, q2=q2, q3=q3, v1=v1, v2=v2, v3=v3,
        a1=a1, a2=a2, a3=a3, x=x, y=y, z=z, vx=vx, vy=vy, vz=vz,
        K=K, Lx=Lx, Ly=Ly, Lz=Lz, Lmag=Lmag,
    )

    # --- Painel de informações ---
    with out_info:
        partes = []
        partes.append(sec_html('Sistema escolhido', 'margin-top:0'))
        partes.append(card(
            f'{cfg["titulo"]} &nbsp;|&nbsp; massa m = {massa:.6g} kg &nbsp;|&nbsp; '
            f'{cfg["F_disp"][0]} = {F1_text.value} &nbsp; '
            f'{cfg["F_disp"][1]} = {F2_text.value} &nbsp; '
            f'{cfg["F_disp"][2]} = {F3_text.value}',
            'rcard'))
        partes.append(sec_html('Condições iniciais e intervalo'))
        ic_rows = ''.join(
            f'<tr><td>{cfg["q_disp"][i]}₀</td><td>{q0[i]:.6g} {up}'
            f'{" (rad)" if cfg["q_kind"][i]=="ang" else ""}</td>'
            f'<td>{cfg["v_disp"][i]}₀</td><td>{v0[i]:.6g} {up}/{ut}</td></tr>'
            for i in range(3)
        )
        partes.append(
            f'<table class="res-t"><tr><th>Coordenada</th><th>Valor</th>'
            f'<th>Velocidade</th><th>Valor</th></tr>{ic_rows}</table>')
        partes.append(card(
            f'Intervalo: t ∈ [{tmin:.6g}, {tmax:.6g}] {ut} &nbsp;|&nbsp; '
            f'método: {metodo} &nbsp;|&nbsp; pontos: {npts} &nbsp;|&nbsp; '
            f'rtol={rtol:g}, atol={atol:g} &nbsp;|&nbsp; '
            f'passos internos do integrador: {len(sol.t)} (nfev={sol.nfev})',
            'bcard'))
        for a in avisos:
            partes.append(card('⚠ ' + a, 'wcard'))
        display(HTML(''.join(partes)))

    # --- Figura A: trajetória 3D ---
    fig_traj = plt.figure(figsize=(7.5, 6.5))
    ax = fig_traj.add_subplot(111, projection='3d')
    ax.plot(x, y, z, color=COR_TRAJ, lw=1.8)
    ax.scatter([x[0]], [y[0]], [z[0]], color='#2f9e44', s=60, label='Início', depthshade=False)
    ax.scatter([x[-1]], [y[-1]], [z[-1]], color='#c92a2a', s=60, label='Fim', depthshade=False)
    ax.set_xlabel(f'x [{up}]')
    ax.set_ylabel(f'y [{up}]')
    ax.set_zlabel(f'z [{up}]')
    ax.set_title('a) Trajetória 3D', fontsize=12, fontweight='bold', color='#0d47a1')
    ax.legend(loc='upper left', fontsize=8)
    fig_traj.tight_layout()
    state['fig_traj'] = fig_traj
    with out_traj:
        display(HTML(fig_para_html(fig_traj, max_width=760)))
    plt.close(fig_traj)

    # --- Figura B: grade 3x3 (posição, velocidade, aceleração) ---
    fig_grid, axs = plt.subplots(3, 3, figsize=(13.5, 9.5), sharex=True)
    dados_q = [q1, q2, q3]
    dados_v = [v1, v2, v3]
    dados_a = [a1, a2, a3]
    letras = ['b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

    for j in range(3):
        unit_q = up if cfg['q_kind'][j] == 'pos' else 'rad'
        axs[0, j].plot(ts, dados_q[j], color=COR_POS, lw=1.8)
        axs[0, j].set_ylabel(f"{cfg['q_disp'][j]}(t)  [{unit_q}]")
        axs[0, j].set_title(f"{letras[j]}) {cfg['q_disp'][j]}(t)", fontsize=10, loc='left', color=COR_POS)

        axs[1, j].plot(ts, dados_v[j], color=COR_VEL, lw=1.8)
        axs[1, j].set_ylabel(f"{cfg['v_disp'][j]}(t)  [{up}/{ut}]")
        axs[1, j].set_title(f"{letras[j+3]}) {cfg['v_disp'][j]}(t)", fontsize=10, loc='left', color=COR_VEL)

        axs[2, j].plot(ts, dados_a[j], color=COR_ACE, lw=1.8)
        axs[2, j].set_ylabel(f"{cfg['a_disp'][j]}(t)  [{up}/{ut}²]")
        axs[2, j].set_title(f"{letras[j+6]}) {cfg['a_disp'][j]}(t)", fontsize=10, loc='left', color=COR_ACE)
        axs[2, j].set_xlabel(f't  [{ut}]')

    for ax in axs.flat:
        ax.grid(True, alpha=0.3)

    fig_grid.suptitle('Posição, velocidade e aceleração (componentes físicas) em função do tempo',
                       fontsize=13, fontweight='bold', color='#0d47a1')
    fig_grid.tight_layout(rect=[0, 0, 1, 0.96])
    state['fig_grid'] = fig_grid
    with out_grid:
        display(HTML(fig_para_html(fig_grid, max_width=1150)))
    plt.close(fig_grid)

    # --- Figura C: gráficos extras (energia cinética e momento angular) ---
    fig_extra, axe = plt.subplots(1, 2, figsize=(12, 4.2))
    axe[0].plot(ts, K, color='#c2255c', lw=1.8)
    axe[0].set_xlabel(f't  [{ut}]')
    axe[0].set_ylabel('K(t)  [J]  (assumindo SI)')
    v_lbl = cfg['v_disp']
    axe[0].set_title(
        f'k) Energia cinética K(t) = ½·m·({v_lbl[0]}²+{v_lbl[1]}²+{v_lbl[2]}²)',
        fontsize=10, loc='left', color='#c2255c')
    axe[0].grid(True, alpha=0.3)

    axe[1].plot(ts, Lmag, color='#5f3dc4', lw=1.8, label='|L(t)|')
    axe[1].plot(ts, Lz, color='#5f3dc4', lw=1.2, ls='--', alpha=0.7, label='L_z(t)')
    axe[1].set_xlabel(f't  [{ut}]')
    axe[1].set_ylabel('Momento angular  [kg·{}²/{}]'.format(up, ut))
    axe[1].set_title('l) Momento angular em relação à origem', fontsize=10, loc='left', color='#5f3dc4')
    axe[1].legend(fontsize=8)
    axe[1].grid(True, alpha=0.3)

    fig_extra.suptitle('Diagnósticos físicos adicionais', fontsize=12, fontweight='bold', color='#0d47a1')
    fig_extra.tight_layout(rect=[0, 0, 1, 0.93])
    state['fig_extra'] = fig_extra
    with out_extra:
        display(HTML(fig_para_html(fig_extra, max_width=1000)))
        display(HTML(card(
            'ℹ Se a força for central (ou nula) e F_θ = F_φ = 0, o momento angular '
            '|L| e L_z tendem a se manter aproximadamente constantes — um bom '
            'critério visual para checar a qualidade da integração numérica.',
            'icard')))
    plt.close(fig_extra)

    with out_export:
        display(HTML(card(
            '✔ Simulação concluída. Use os botões acima para exportar os resultados.',
            'gcard')))

# ----------------------- Exportação -----------------------

def _baixar(caminho):
    if IN_COLAB:
        colab_files.download(caminho)
    else:
        display(HTML(card(f'Arquivo salvo em: <code>{caminho}</code>', 'icard')))


def _monta_df_dados():
    cfg = COORD_DEFS[state['coord_name']]
    up, ut = state['up'], state['ut']
    unit_q = [up if k == 'pos' else 'rad' for k in cfg['q_kind']]
    cols = {f"t [{ut}]": state['ts']}
    for i in range(3):
        cols[f"{cfg['q_disp'][i]}(t) [{unit_q[i]}]"] = state[f'q{i+1}']
    for i in range(3):
        cols[f"{cfg['v_disp'][i]}(t) [{up}/{ut}]"] = state[f'v{i+1}']
    for i in range(3):
        cols[f"{cfg['a_disp'][i]}(t) [{up}/{ut}²]"] = state[f'a{i+1}']
    cols[f"x(t) [{up}]"] = state['x']
    cols[f"y(t) [{up}]"] = state['y']
    cols[f"z(t) [{up}]"] = state['z']
    cols["K(t) [J, assumindo SI]"] = state['K']
    cols["L_x(t)"] = state['Lx']
    cols["L_y(t)"] = state['Ly']
    cols["L_z(t)"] = state['Lz']
    cols["|L(t)|"] = state['Lmag']
    return pd.DataFrame(cols)


def _monta_df_parametros():
    cfg = COORD_DEFS[state['coord_name']]
    rows = [
        ('Sistema de coordenadas', cfg['titulo']),
        ('Massa m (kg)', state['massa']),
        (f"{cfg['F_disp'][0]}", state['F_exprs'][0]),
        (f"{cfg['F_disp'][1]}", state['F_exprs'][1]),
        (f"{cfg['F_disp'][2]}", state['F_exprs'][2]),
        (f"{cfg['q_disp'][0]}₀", state['q0'][0]),
        (f"{cfg['q_disp'][1]}₀", state['q0'][1]),
        (f"{cfg['q_disp'][2]}₀", state['q0'][2]),
        (f"{cfg['v_disp'][0]}₀", state['v0'][0]),
        (f"{cfg['v_disp'][1]}₀", state['v0'][1]),
        (f"{cfg['v_disp'][2]}₀", state['v0'][2]),
        ('t mínimo', state['tmin']),
        ('t máximo', state['tmax']),
        ('Unidade de posição', state['up']),
        ('Unidade de tempo', state['ut']),
        ('Método do integrador', state['metodo']),
        ('rtol', state['rtol']),
        ('atol', state['atol']),
        ('Nº de pontos', state['npts']),
    ]
    return pd.DataFrame(rows, columns=['Parâmetro', 'Valor'])


def on_export_png(btn=None):
    with out_export:
        if 'fig_traj' not in state:
            display(HTML(card('⚠ Execute a simulação antes de exportar.', 'wcard')))
            return
        caminhos = []
        for nome, fig in (('trajetoria_3d', state['fig_traj']),
                           ('posicao_velocidade_aceleracao', state['fig_grid']),
                           ('diagnosticos', state['fig_extra'])):
            caminho = f'/content/dinamica3d_{nome}.png'
            fig.savefig(caminho, dpi=150, bbox_inches='tight')
            caminhos.append(caminho)
        display(HTML(card('✔ Gráficos exportados (3 arquivos PNG).', 'gcard')))
        for c in caminhos:
            _baixar(c)


def on_export_xlsx(btn=None):
    with out_export:
        if 'ts' not in state:
            display(HTML(card('⚠ Execute a simulação antes de exportar.', 'wcard')))
            return
        caminho = '/content/dinamica3d_resultados.xlsx'
        with pd.ExcelWriter(caminho) as writer:
            _monta_df_dados().to_excel(writer, sheet_name='Dados_t_q_v_a', index=False)
            _monta_df_parametros().to_excel(writer, sheet_name='Parametros', index=False)
        display(HTML(card(f'✔ Planilha exportada: <code>{caminho}</code>', 'gcard')))
        _baixar(caminho)


def on_export_csv(btn=None):
    with out_export:
        if 'ts' not in state:
            display(HTML(card('⚠ Execute a simulação antes de exportar.', 'wcard')))
            return
        caminho = '/content/dinamica3d_resultados_csv.zip'
        with zipfile.ZipFile(caminho, 'w') as zf:
            zf.writestr('Dados_t_q_v_a.csv', _monta_df_dados().to_csv(index=False))
            zf.writestr('Parametros.csv', _monta_df_parametros().to_csv(index=False))
        display(HTML(card(f'✔ Dados exportados (ZIP com CSVs): <code>{caminho}</code>', 'gcard')))
        _baixar(caminho)


simular_btn.on_click(on_simular_clicked)
export_png_btn.on_click(on_export_png)
export_xlsx_btn.on_click(on_export_xlsx)
export_csv_btn.on_click(on_export_csv)


# ----------------------- Interface -----------------------

ui = w.VBox([
    cabecalho,

    titulo_secao('⚙️ Partícula, sistema de coordenadas e força'),
    w.HBox([massa_text, coord_dropdown]),
    obs_coord_html,
    out_eqs,
    F1_text,
    F2_text,
    F3_text,

    titulo_secao('🎯 Condições iniciais e intervalo de simulação'),
    w.HBox([tmin_text, tmax_text]),
    w.HBox([q1_0, q2_0, q3_0]),
    w.HBox([v1_0, v2_0, v3_0]),
    w.HBox([unidade_pos, unidade_tempo]),
    unidades_obs,

    opcoes_avancadas,
    simular_btn,

    titulo_secao('📋 Informações da simulação'),
    out_info,

    titulo_secao('📊 a) Trajetória 3D'),
    out_traj,

    titulo_secao('📊 b–j) Posição, velocidade e aceleração vs. tempo'),
    out_grid,

    titulo_secao('📊 k–l) Diagnósticos físicos adicionais'),
    out_extra,

    titulo_secao('💾 Exportar resultados'),
    w.HBox([export_png_btn, export_xlsx_btn, export_csv_btn]),
    out_export,
])

display(ui)
```
